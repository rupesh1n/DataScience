```python
#z-test
'''A telecom company claims that the average
monthly data usage of its customers is 50GB.
A data analyst suspects that the average usage 
has increased due to increased streaming services.

Sample of 10 customers
population std=5GB
significance level=0.05'''
```

```python
Ha='''Average usage > 50GB'''
H0='''Average usage =50GB'''
```

```python
import numpy as np
from scipy.stats import norm
data=[52,55,49,60,58,54,53,57,51,56]
mu=50
sigma=5
alpha=0.05
n=len(data)
sample_mean=np.mean(data)
z=(sample_mean-mu)/(sigma/np.sqrt(n))
p_value=1-norm.cdf(z)
print("sample_mean:",sample_mean)
print("z statistic:",z)
print("p value:",p_value)
if(p_value<=alpha):
    print("Reject Null hypothesis")
else:
    print("Accept Null hypothesis")
```

```python
'''A school claims that the average score of students
is 70.
A teacher collects marks of 8 students from a class
after using a new teaching technique.
Marks obtained:72,75,78,69,71,73,76.
Is the average score significantly different from 70?'''
```

```python
from scipy import stats
marks=[72,75,78,69,71,73,76]
population_mean=70
test_stat,p=stats.ttest_1samp(marks,population_mean)
print("Test statistic:",test_stat)
print("p-value:",p)
if(p<=alpha):
    print("Reject Null hypothesis")
else:
    print("Accept Null hypothesis")
```

```python
#independent t test(two sample t test)
'''compare the means of two independent groups'''
```

```python
'''Two groups of students:
Group        Teaching method
Group 1        classroom
Group 2         Online

classroom      online
72               75
70               78
68                74
74               80
69                77'''
```

```python
classroom=[72,70,68,74,69]
online=[75,78,74,80,77]
```

```python
test,p=stats.ttest_ind(classroom,online)
print("Test statistic:",test)
print("p-value:",p)
if(p<=alpha):
    print("Reject H0 ")
else:
    print("Accept H0")
```

```python
'''ANOVA(Analysis of variance)
It is used when we want to compare the means of
3 or more groups to see if they are significantly different.'''
```

```python
'''ANOVA compares two types of variation:
1.Variation between groups--how much the group averages
differ
2.Variation within groups --- how much the values inside
each group differ'''
```

```python
a=[78,82,85]
b=[88,90,92]
c=[70,72,68]
test,p=stats.f_oneway(a,b,c)
if(p<=alpha):
    print("Reject H0")
else:
    print("Accept H0")
```
