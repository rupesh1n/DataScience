```python
#Machine Learning
'''Machine learning is a field of Artificial Intelligence
that enables computers to learn patterns from data and 
make decisions or predictions without being explicitly
programmed.'''
```

```python
#Types of machine learning:
#1.Supervised learning:
'''The model learns from labeled data.
Labeled data=Input+output variables
Example:Data set for house prediction:
Size(sqft)   Price
1000          50L
1200          60L
1500          75L'''
```

```python
#Types of supervised learning:
'''1.classification:
Output is categorical
Eg:
spam/not spam
Disease/No disease
Customer churn/no churn'''
```

```python
'''2.Regression:
Output is continuous numeric value.
Eg:
House price prediction
Temperature prediction
Sales forecasting'''
```

```python
'''output variable:y
Target variable/dependent variable
Input variable:x
Independent variables'''
```

```python
#Unsupervised learning:
'''In unsupervised learning,the data set does not
have labels.
Eg:customer data
customer    Age     spending
c1          22        5000
c2          23       5200
c3           45       20000
c4         46           21000

The algorithm groups data.
The process of grouping data is called clustering.'''
```

```python
#Reinforcement learning:
'''Model learns by interacting with an environment.
The system learns using rewards and penalties.'''
```

```python
#training
'''Is the process where a machine learning model
learns patterns from a dataset'''
```

```python
#testing
'''Is the process of evaluating how well
the trained model works on new,unseen data.
Divide the dataset into two parts:
1.Training data --- used to train the model
2.Testing data --- used to evaluate the model'''
```

```python
#Overfitting:
'''occurs when a model learns the training data too well,
including noise and small details.
Causes:
Too many features
Too many trainig iterations
very complex model
small dataset'''
```

```python
#Underfitting:
'''Underfitting occurs when the model fails to learn
the patterns in the data.
The model is too simple to capture the relationship 
between the input and output.
Causes:
very simple model
Too few features
Not enough training'''
```

```python
#R square:
'''How much of real outcome can my model successfully
explain?
R square value      Meaning
1                    Perfect prediction
0                    Same as guessing average
<0                    worse than average

Eg:R2=0.8 --- Model explains 80% of variation
R2=0.3 --- weak model
Higher R2 better model'''
```
