```python
class Customer:
    def __init__(self, name: str, panNo: str, emailId: str, salary: int):
        # Private attributes
        self.__name = name
        self.__panNo = panNo
        self.__emailId = emailId
        self.__salary = salary

    # name property
    @property
    def name(self) -> str:
        return self.__name

    @name.setter
    def name(self, value: str) -> None:
        self.__name = value

    # panNo property
    @property
    def panNo(self) -> str:
        return self.__panNo

    @panNo.setter
    def panNo(self, value: str) -> None:
        self.__panNo = value

    # emailId property
    @property
    def emailId(self) -> str:
        return self.__emailId

    @emailId.setter
    def emailId(self, value: str) -> None:
        self.__emailId = value

    # salary property
    @property
    def salary(self) -> int:
        return self.__salary

    @salary.setter
    def salary(self, value: int) -> None:
        self.__salary = int(value)

    def __eq__(self, other) -> bool:
        """
         if both emailId AND panNo match exactly then object will be wqual
        """
        if not isinstance(other, Customer):
            return NotImplemented
        return (self.emailId == other.emailId) and (self.panNo == other.panNo)


class Main:
    @staticmethod
    def main():
        # Read first customer
        name1 = input("Enter first customer name: ")
        pan1 = input("Enter first customer panno: ")
        email1 = input("Enter first customer emailid: ")
        salary1 = int(input("Enter first customer salary: "))

        # Read second customer
        name2 = input("Enter second customer name: ")
        pan2 = input("Enter second customer panno: ")
        email2 = input("Enter second customer emailid: ")
        salary2 = int(input("Enter second customer salary: "))

        # Create Customer objects
        c1 = Customer(name1, pan1, email1, salary1)
        c2 = Customer(name2, pan2, email2, salary2)

        # Compare and print result
        if c1 == c2:
            print("Both the objects are equal.")
        else:
            print("Both the objects are not equal.")


if __name__ == "__main__":
    Main.main()
```

```python
class Ticket:
    def __init__(self, ticketId: int, passengerName: str, gender: str, address: str):
        self.__ticketId = ticketId
        self.__passengerName = passengerName
        self.__gender = gender
        self.__address = address

    # ticketId property
    @property
    def ticketId(self) -> int:
        return self.__ticketId

    @ticketId.setter
    def ticketId(self, value: int) -> None:
        if isinstance(value, int):
            self.__ticketId = value
        else:
            try:
                self.__ticketId = int(value)
            except Exception as e:
                raise ValueError("ticketId must be an integer") from e

    # passengerName property
    @property
    def passengerName(self) -> str:
        return self.__passengerName

    @passengerName.setter
    def passengerName(self, value: str) -> None:
        self.__passengerName = value

    # gender property
    @property
    def gender(self) -> str:
        return self.__gender

    @gender.setter
    def gender(self, value: str) -> None:
        self.__gender = value

    # address property
    @property
    def address(self) -> str:
        return self.__address

    @address.setter
    def address(self, value: str) -> None:
        self.__address = value

    def __str__(self) -> str:
        # Dot-matrix style single-line print
        return (
            f"ticketId:{self.ticketId}/"
            f"passengerName:{self.passengerName}/"
            f"gender:{self.gender}/"
            f"address:{self.address}"
        )


class UserInterface:
    @staticmethod
    def main():
        # Read number of passengers (prompt format as per sample)
        try:
            n_str = input("Enter the number of passengers:")
            n = int(n_str.strip())
        except ValueError:
            # If the input is invalid, treat as zero or print message; we'll print and exit.
            print("Invalid number of passengers.")
            return

        tickets = []

        for i in range(1, n + 1):
            print(f"\nPassenger {i}:")
            # Follow the sample prompts exactly (spacing after colon as seen in sample)
            # Note: sample mixes with/without spaces; we keep consistent with a space after colon
            # for field lines, while the first line (number of passengers) had no space.
            ticketId_str = input("Enter the ticketId: ").strip()
            passengerName = input("Enter the passengerName:").strip()
            gender = input("Enter the gender:").strip()
            address = input("Enter the address:").strip()

            try:
                ticketId = int(ticketId_str)
            except ValueError:
                print("Invalid ticketId. It must be an integer.")
                return

            tickets.append(Ticket(ticketId, passengerName, gender, address))

        # Blank line before output, as in the sample
        print()
        for t in tickets:
            print(str(t))


if __name__ == "__main__":
    UserInterface.main()
```

```python
from abc import ABC, abstractmethod


class Student(ABC):
    def __init__(self, studentId: int, studentName: str, department: str,
                 gender: str, category: str, collegeFee: float):
        self._studentId = int(studentId)
        self._studentName = studentName
        self._department = department
        self._gender = gender
        self._category = category
        self._collegeFee = float(collegeFee)

    # ---- Properties (public getters/setters) for all protected attributes ----
    @property
    def studentId(self) -> int:
        return self._studentId

    @studentId.setter
    def studentId(self, value: int) -> None:
        self._studentId = int(value)

    @property
    def studentName(self) -> str:
        return self._studentName

    @studentName.setter
    def studentName(self, value: str) -> None:
        self._studentName = value

    @property
    def department(self) -> str:
        return self._department

    @department.setter
    def department(self, value: str) -> None:
        self._department = value

    @property
    def gender(self) -> str:
        return self._gender

    @gender.setter
    def gender(self, value: str) -> None:
        self._gender = value

    @property
    def category(self) -> str:
        return self._category

    @category.setter
    def category(self, value: str) -> None:
        self._category = value

    @property
    def collegeFee(self) -> float:
        return self._collegeFee

    @collegeFee.setter
    def collegeFee(self, value: float) -> None:
        self._collegeFee = float(value)

    # ---- Abstract method ----
    @abstractmethod
    def calculate_total_fee(self) -> float:
        pass


class Hosteller(Student):
    def __init__(self, studentId: int, studentName: str, department: str, gender: str,
                 category: str, collegeFee: float, roomNumber: int,
                 blockName: str, roomType: str):
        super().__init__(studentId, studentName, department, gender, category, collegeFee)
        self.__roomNumber = int(roomNumber)
        self.__blockName = blockName          # case-sensitive (A/B/C expected)
        self.__roomType = roomType            # case-sensitive ("AC" or "Non-AC" expected)

    # ---- Properties for private attributes ----
    @property
    def roomNumber(self) -> int:
        return self.__roomNumber

    @roomNumber.setter
    def roomNumber(self, value: int) -> None:
        self.__roomNumber = int(value)

    @property
    def blockName(self) -> str:
        return self.__blockName

    @blockName.setter
    def blockName(self, value: str) -> None:
        self.__blockName = value

    @property
    def roomType(self) -> str:
        return self.__roomType

    @roomType.setter
    def roomType(self, value: str) -> None:
        self.__roomType = value

    def calculate_total_fee(self) -> float:
        # Hostel fee mapping (common to AC and Non-AC)
        hostel_fee_map = {
            "A": 60000,
            "B": 50000,
            "C": 40000
        }
        # AC add-on mapping
        ac_addon_map = {
            "A": 8000,
            "B": 5000,
            "C": 2500
        }

        block = self.blockName  # case-sensitive
        if block not in hostel_fee_map:
            raise ValueError("Invalid block name. Expected one of: A, B, C.")

        total = self.collegeFee + hostel_fee_map[block]

        # Only add AC add-on if roomType exactly equals "AC"
        if self.roomType == "AC":
            total += ac_addon_map[block]

        return float(total)


class DayScholar(Student):
    def __init__(self, studentId: int, studentName: str, department: str, gender: str,
                 category: str, collegeFee: float, busNumber: int, distance: float):
        super().__init__(studentId, studentName, department, gender, category, collegeFee)
        self.__busNumber = int(busNumber)
        self.__distance = float(distance)

    # ---- Properties for private attributes ----
    @property
    def busNumber(self) -> int:
        return self.__busNumber

    @busNumber.setter
    def busNumber(self, value: int) -> None:
        self.__busNumber = int(value)

    @property
    def distance(self) -> float:
        return self.__distance

    @distance.setter
    def distance(self, value: float) -> None:
        self.__distance = float(value)

    def calculate_total_fee(self) -> float:
        d = self.distance
        if not (0 < d <= 40):
            raise ValueError("Distance must be in the range (0, 40].")

        if d <= 10:
            bus_fee = 6000
        elif d <= 20:
            bus_fee = 12000
        elif d <= 30:
            bus_fee = 20000
        else:  # d <= 40
            bus_fee = 28000

        return float(self.collegeFee + bus_fee)


class UserInterface:
    @staticmethod
    def main():
        # ---- Read common inputs ----
        student_id = int(input("Enter student id: ").strip())
        student_name = input("Enter student name: ").strip()
        department = input("Enter department: ").strip()
        gender = input("Enter gender: ").strip()
        category = input("Enter category: ").strip()  # Expect "Hosteller" or "DayScholar"

        student_obj = None

        if category == "Hosteller":
            room_number = int(input("Enter room number: ").strip())
            block_name = input("Enter block name: ").strip()     # case-sensitive
            room_type = input("Enter room type: ").strip()       # "AC" or "Non-AC" (case-sensitive)
            college_fee = float(input("Enter college fee: ").strip())

            student_obj = Hosteller(
                studentId=student_id,
                studentName=student_name,
                department=department,
                gender=gender,
                category=category,
                collegeFee=college_fee,
                roomNumber=room_number,
                blockName=block_name,
                roomType=room_type
            )

        elif category == "DayScholar":
            bus_number = int(input("Enter bus number: ").strip())
            distance = float(input("Enter distance: ").strip())
            college_fee = float(input("Enter college fee: ").strip())

            student_obj = DayScholar(
                studentId=student_id,
                studentName=student_name,
                department=department,
                gender=gender,
                category=category,
                collegeFee=college_fee,
                busNumber=bus_number,
                distance=distance
            )
        else:
            # If needed, you can handle invalid category differently as per platform constraints
            raise ValueError('Invalid category. Expected "Hosteller" or "DayScholar".')

        total = student_obj.calculate_total_fee()
        print(float(total))


if __name__ == "__main__":
    UserInterface.main()
```
