Why CSV and Excel?
CSV (Comma-Separated Values):
	• Universal format supported by virtually all data tools
	• Lightweight and fast to read/write
	• Human-readable plain text
	• Ideal for large datasets
Excel (.xlsx, .xls):
	• Widely used in business environments
	• Supports multiple sheets in one file
	• Preserves formatting, formulas, and data types
	• Familiar interface for non-technical users

#### CSV Files - Reading and Writing
Reading CSV Files with pd.read_csv()

sep (separator): Delimiter character (default is comma)
#### Tab-separated values
df = pd.read_csv('data.tsv', sep='\t')
#### Pipe-separated
df = pd.read_csv('data.txt', sep='|')
#### Semicolon-separated (common in European formats)
df = pd.read_csv('data.csv', sep=';')
header: Row number to use as column names

#### First row as header (default)
df = pd.read_csv('data.csv', header=0)
#### No header in file, pandas assigns column names
df = pd.read_csv('data.csv', header=None)
#### Use second row as header
df = pd.read_csv('data.csv', header=1)
names: Custom column names
#### Provide your own column names
column_names = ['ID', 'Name', 'Age', 'Salary']
df = pd.read_csv('data.csv', names=column_names, header=0)

encoding: File encoding
#### UTF-8 (default)
df = pd.read_csv('data.csv', encoding='utf-8')
#### Latin-1 for European characters
df = pd.read_csv('data.csv', encoding='latin-1')
#### Windows encoding
df = pd.read_csv('data.csv', encoding='cp1252')

```python
import csv

# ---------------------------
# 1. CREATE & WRITE TO CSV
# ---------------------------
with open("Person.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)

    # Write header
    writer.writerow(["ID", "Name", "Age"])

    # Write data rows
    writer.writerow([1, "Vikas", 35])
    writer.writerow([2, "Rahul", 28])
    writer.writerow([3, "Meera", 30])

print("Data written to Person.csv\n")

# ---------------------------
# 2. READ THE CSV FILE
# ---------------------------
persons = []

with open("Person.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    header = next(reader)   # Skip header row

    for row in reader:
        persons.append(row)

# ---------------------------
# 3. FORMAT & PRINT DATA
# ---------------------------
print("Formatted Output:\n")
print(f"{'ID':<5} {'Name':<10} {'Age':<5}")
print("-" * 25)

for p in persons:
    print(f"{p[0]:<5} {p[1]:<10} {p[2]:<5}")
```

```python
import pandas as pd

# Read data
df_stocks = pd.read_csv('Stock.csv')

# 1. Clean the Growth (%) column
df_stocks['Growth (%)'] = df_stocks['Growth (%)'].str.replace('%', '').astype(float)

# 2. NEW: Clean the Price columns (Remove ₹ and any commas, then convert to float)
# We use .replace to remove symbols so we can treat them as numbers
df_stocks['Target Price (₹)'] = df_stocks['Target Price (₹)'].str.replace('₹', '').str.replace(',', '').astype(float)
df_stocks['Current Price (₹)'] = df_stocks['Current Price (₹)'].str.replace('₹', '').str.replace(',', '').astype(float)

# 3. Calculate actual growth amount (Now this works because both are floats!)
df_stocks['Growth Amount (₹)'] = df_stocks['Target Price (₹)'] - df_stocks['Current Price (₹)']

# Display cleaned data
print("Cleaned Stock Data:")
print(df_stocks.head())

# Find stocks with highest growth potential
print("\nTop 5 Stocks by Growth Percentage:")
top_growth = df_stocks.nlargest(5, 'Growth (%)')
print(top_growth[['Script Name', 'Stock Symbol', 'Growth (%)', 'Sector']])
```

```python
#Practical Example 2: Advanced CSV Writing
import pandas as pd
from datetime import datetime
# Create DataFrame with various data types
data = {
    'transaction_id': [1001, 1002, 1003, 1004],
    'date': pd.to_datetime(['2024-01-15', '2024-01-16', '2024-01-17', '2024-01-18']),
    'product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor'],
    'quantity': [2, 5, 3, 1],
    'price': [1299.99, 25.50, 79.99, 349.00],
    'total': [2599.98, 127.50, 239.97, 349.00],
    'notes': ['Express delivery', None, 'Gift wrap', 'Bulk order']
}
df = pd.DataFrame(data)
# Write with specific formatting
df.to_csv(
    'sales_export.csv',
    index=False,                          # No row index
    columns=['date', 'product', 'quantity', 'total'],  # Select columns
    sep=',',                              # Comma separator
    na_rep='N/A',                         # Missing values as 'N/A'
    float_format='%.2f',                  # 2 decimal places for floats
    date_format='%Y-%m-%d',              # Date format
    encoding='utf-8-sig',                 # Excel-friendly encoding
    quoting=1                             # Quote all non-numeric
)
print("CSV file created successfully!")
```

#### Reading JSON Files using json.load() and json.loads()

Python's json module provides two main methods for reading JSON:
json.loads() - Loads JSON from a string.

json.load() - Loads JSON from a file

Basic Example with json.loads():


```python
import json
# JSON as a string
json_string = '{"name": "Bob", "age": 25, "city": "New York"}'
# Parse JSON string

data = json.loads(json_string)
print(data)  # Output: {'name': 'Bob', 'age': 25, 'city': 'New York'}
print(type(data))  # Output: <class 'dict'>
```

```python
import json

with open("payroll.json", encoding="utf-8") as f:
    data = json.load(f)

print(f"Payroll: {data['pay_period']['month']} {data['pay_period']['year']}")
print("EmpID  Name              Location     Gross    Deductions  Net")
print("-"*65)

for emp in data["employees"]:
    pid = emp["employee_id"]
    name = emp["personal"]["name"]
    loc = emp["employment"]["location"]
    gross = emp["current_month"]["earnings"]["gross"]
    ded = emp["current_month"]["deductions"]["total_deductions"]
    net = emp["current_month"]["net_pay"]
    print(f"{pid:<6} {name:<17} {loc:<12} {gross:>8,.2f} {ded:>11,.2f} {net:>8,.0f}")
```

```python
#Compute aggregates by location (e.g., Bengaluru vs others)


from collections import defaultdict
import json

with open("payroll.json", encoding="utf-8") as f:
    data = json.load(f)

agg = defaultdict(lambda: {"gross": 0.0, "deductions": 0.0, "net": 0.0, "count": 0})
for emp in data["employees"]:
    city = emp["employment"]["location"]
    cm = emp["current_month"]
    agg[city]["gross"] += cm["earnings"]["gross"]
    agg[city]["deductions"] += cm["deductions"]["total_deductions"]
    agg[city]["net"] += cm["net_pay"]
    agg[city]["count"] += 1


for city, val in agg.items():
    print(f"{city}: employees={val['count']}, avg_net={val['net']/val['count']:.2f}")
```

```python
# Recalculate PF (employee/employer) from basic and verify

import json

with open("payroll.json", encoding="utf-8") as f:
    data = json.load(f)

MISMATCHES = []
for e in data["employees"]:
    basic = e["pay_structure"]["components_monthly"]["basic"]
    expected_pf = round(basic * 0.12, 2)
    pf_emp = e["current_month"]["deductions"]["pf_employee"]
    pf_er = e["pay_structure"]["employer_contributions"]["pf"]
    if abs(pf_emp - expected_pf) > 0.01 or abs(pf_er - expected_pf) > 0.01:
        MISMATCHES.append((e["employee_id"], e["personal"]["name"], pf_emp, pf_er, expected_pf))

    print("PF check:", "OK" if not MISMATCHES else MISMATCHES)
```

```python
#Export a compact payslip CSV

import json, csv

with open("payroll.json", encoding="utf-8") as f:
 data = json.load(f)

fields = ["employee_id", "name", "location", "gross", "deductions", "net", "bank_ifsc"]
rows = []
for e in data["employees"]:
    rows.append({
    "employee_id": e["employee_id"],
    "name": e["personal"]["name"],
    "location": e["employment"]["location"],
    "gross": e["current_month"]["earnings"]["gross"],
    "deductions": e["current_month"]["deductions"]["total_deductions"],
    "net": e["current_month"]["net_pay"],
    "bank_ifsc": e["bank"]["ifsc"]
    })

with open("payslips.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=fields)
    writer.writeheader()
    writer.writerows(rows)

print("Created payslips.csv")
```
