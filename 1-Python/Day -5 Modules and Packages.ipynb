```python

# use_fibo.py
import fibo

print("Fibonacci up to 30 (printed by fib):")
fibo.fib(30)   # prints numbers

print("Fibonacci up to 100 (captured by fib2):")
series = fibo.fib2(100)
print(series)  # prints the list
```

```python

# use_fibo_selective.py
from fibo import fib, fib2

fib(40)
print(fib2(150))
```

```python

# use_fibo_alias.py
import fibo as fb

fb.fib(60)
print(fb.fib2(300))
```

```python

# use_fibo_alias_selective.py
from fibo import fib as print_fib, fib2 as list_fib

print_fib(55)
print(list_fib(233))
```

```python

# use_calculator.py
import calculator

print("Addition:", calculator.add(10, 5))
print("Subtraction:", calculator.subtract(10, 5))
print("Multiplication:", calculator.multiply(10, 5))

print("True division:", calculator.divide(7, 2))                  # 3.5
print("Integer division:", calculator.divide(7, 2, integer=True))  # 3

print("Modulus:", calculator.modulus(7, 2))
```
