```python
import math

class Triangle:
    def __init__(self, side1, side2, side3):
        self.side1 = side1
        self.side2 = side2
        self.side3 = side3
        if not self.is_valid_triangle():
            raise ValueError("Sides do not form a valid triangle")

    def is_valid_triangle(self):
        a, b, c = self.side1, self.side2, self.side3
        return a + b > c and a + c > b and b + c > a

    def perimeter(self):
        return self.side1 + self.side2 + self.side3

    def area(self):
        # Heron's formula
        s = self.perimeter() / 2
        return math.sqrt(s * (s - self.side1) * (s - self.side2) * (s - self.side3))

    def __str__(self):
        return f"Triangle({self.side1}, {self.side2}, {self.side3})"


class Equilateral(Triangle):
    def __init__(self, side):
        super().__init__(side, side, side)
        # Redundant check, but explicit for clarity
        if not (self.side1 == self.side2 == self.side3):
            raise ValueError("Equilateral triangle must have all sides equal")

    def area(self):
        # Special formula for equilateral triangle
        return (math.sqrt(3) / 4) * self.side1 ** 2

    def __str__(self):
        return f"Equilateral({self.side1})"


class RTriangle(Triangle):  # Right Triangle
    def __init__(self, leg1, leg2, hypotenuse=None):
        if hypotenuse is None:
            hypotenuse = math.sqrt(leg1**2 + leg2**2)
        super().__init__(leg1, leg2, hypotenuse)
        # Validate it's a right triangle using Pythagoras
        sides = sorted([self.side1, self.side2, self.side3])
        if not math.isclose(sides[0]**2 + sides[1]**2, sides[2]**2):
            raise ValueError("Sides do not form a right triangle")

    def area(self):
        # Area = 1/2 * leg1 * leg2
        legs = [self.side1, self.side2, self.side3]
        legs.sort()
        return 0.5 * legs[0] * legs[1]

    def __str__(self):
        return f"RTriangle({self.side1}, {self.side2}, {self.side3})"


class Isosceles(Triangle):  # Fixed spelling: Isoceles -> Isosceles
    def __init__(self, equal_side, base):
        super().__init__(equal_side, equal_side, base)
        if not (self.side1 == self.side2 or self.side2 == self.side3 or self.side1 == self.side3):
            raise ValueError("Isosceles triangle must have at least two equal sides")

    def area(self):
        # Height by Pythagorean theorem: h = sqrt(a^2 - (b/2)^2)
        a = self.side1  # equal side
        b = self.side3  # base
        height = math.sqrt(a**2 - (b/2)**2)
        return 0.5 * b * height

    def __str__(self):
        return f"Isosceles({self.side1}, {self.side2}, {self.side3})"

# Equilateral
e = Equilateral(5)
print(e)          # Equilateral(5)
print(e.area())   # ~10.825
# =====================================================================
# Right Triangle
r = RTriangle(3, 4)  # auto computes hypotenuse = 5
print(r)          # RTriangle(3, 4, 5.0)
print(r.area())   # 6.0

# Isosceles
i = Isosceles(5, 6)
print(i)          # Isosceles(5, 5, 6)
print(i.area())   # 12.0

# Generic Triangle
t = Triangle(3, 4, 5)
print(t.area())   # 6.0
```

The @property decorator in Python is used to define getter methods that can be accessed like attributes, without explicitly calling them as methods. It’s part of Python’s property management system that allows you to encapsulate instance variables and control access to them in a clean, Pythonic way.

 Benefits of Using @property

Encapsulation: You can hide internal implementation and expose a clean interface.

Read-only attributes: You can make attributes that are computed and not directly settable.

Validation logic: You can add checks when setting values using @<property>.setter.

```python
class Temperature:
    """Temperature stored internally in Celsius with convenient properties."""

    def __init__(self, celsius: float):
        # Use the property to leverage validation
        self.celsius = celsius

    @property
    def celsius(self) -> float:
        return self._celsius

    @celsius.setter
    def celsius(self, value: float) -> None:
        if value < -273.15:
            raise ValueError("Temperature can't be below absolute zero!")
        self._celsius = float(value)

    @property
    def fahrenheit(self) -> float:
        return (self._celsius * 9 / 5) + 32

    @fahrenheit.setter
    def fahrenheit(self, value: float) -> None:
        # Convert F -> C, then reuse Celsius validation
        c = (float(value) - 32) * 5 / 9
        self.celsius = c

    @property
    def kelvin(self) -> float:
        return self._celsius + 273.15

    def __repr__(self) -> str:
        return f"Temperature(celsius={self._celsius:.2f})"

    def __str__(self) -> str:
        return f"{self._celsius:.2f} °C / {self.fahrenheit:.2f} °F / {self.kelvin:.2f} K"


def main():
    print("=== Temperature Demo ===")
    # Create a temperature in Celsius
    t = Temperature(25)
    print("Initial:", t)  # uses __str__

    # Read properties
    print("Celsius:", t.celsius)
    print("Fahrenheit:", t.fahrenheit)
    print("Kelvin:", t.kelvin)

    # Set in Celsius
    t.celsius = 0
    print("\nAfter setting to 0 °C:", t)

    # Set in Fahrenheit (will update internal Celsius)
    t.fahrenheit = 212
    print("\nAfter setting to 212 °F (should be 100 °C):", t)

    # Demonstrate validation (below absolute zero)
    try:
        t.celsius = -300
    except ValueError as e:
        print("\nAttempt to set -300 °C:", e)

    # Another demo: set Fahrenheit to absolute zero
    try:
        t.fahrenheit = -459.67  # -273.15 °C
        print("\nSet to absolute zero via °F:", t)
    except ValueError as e:
        print("Error:", e)


if __name__ == "__main__":
    main()
```

### Class variable

A class variable is a variable that is shared across all instances of a class.
It belongs to the class itself, not to any individual object. This means every object of that class accesses and modifies the same shared value.
Key Characteristics

Defined inside the class, but outside any method.
Shared among all instances of the class.
Accessed using:

ClassName.variable_name
instance.variable_name (not recommended for assignment since it creates an instance variable)

Why Use Class Variables?

To maintain common data such as:

Counters (e.g., total number of objects created)
Constants
Shared state or configuration across all objects

```python
class Robot:
    """Represents a robot, with a name."""
    # A class variable, counting the number of robots
    population = 0
    def __init__(self, name):
        """Initializes the data."""
        self.name = name
        print("(Initializing {0}and {1})".format(*self.name.split()))
        # When this person is created, the robot
        # adds to the population
        Robot.population += 1

    def die(self):
        """I am dying."""
        print("{0} is being destroyed!".format(self.name))
        Robot.population -= 1
        if Robot.population == 0:
            print("{0} was the last one.".format(self.name))
        else:
            print("There are still {0:d} robots working.".format(
            Robot.population))

    def sayHi(self):
        """Greeting by the robot.    Yeah, they can do that."""
        print("Greetings, my masters call me {0} .".format(self.name))

    @classmethod
    def howMany(cls):
        """Prints the current population."""
        print("We have {0:d} robots.".format(cls.population))

print(Robot.__doc__)
droid1 = Robot("R2-D2 First Root")
print("**********************")
print(droid1.__init__.__doc__)
      
droid1.sayHi()
Robot.howMany()
droid2 = Robot("C-3PO Scond")
droid2.sayHi()
Robot.howMany()
print("\nRobots can do some work here.\n")
print("Robots have finished their work. So let's destroy them.")

droid1.die()
droid2.die()
Robot.howMany()
```

```python
class Rational(object):
    def __init__(self, numerator=0, denominator=1):
        if denominator == 0:
            raise ZeroDivisionError("Denominator cannot be zero.")
        self.numer = int(numerator)
        self.denom = int(denominator)
        self.fix_sign()
        self.reduce()

    # ---- Arithmetic operators ----
    def __add__(self, rhs):  # self and rhs are Rational objects
        n1 = self.numer      # numerator of the self object
        d1 = self.denom      # denominator of the self object
        n2 = rhs.numer       # numerator of the rhs object
        d2 = rhs.denom       # denominator of the rhs object
        r = Rational(d2 * n1 + d1 * n2, d1 * d2)
        return r

    def __sub__(self, rhs):
        n1 = self.numer
        d1 = self.denom
        n2 = rhs.numer
        d2 = rhs.denom
        r = Rational(d2 * n1 - d1 * n2, d1 * d2)
        return r

    def __mul__(self, rhs):
        return Rational(self.numer * rhs.numer, self.denom * rhs.denom)

    def __truediv__(self, rhs):
        if rhs.numer == 0:
            raise ZeroDivisionError("Division by zero rational.")
        return Rational(self.numer * rhs.denom, self.denom * rhs.numer)

    def __neg__(self):
        return Rational(-self.numer, self.denom)

    def __abs__(self):
        return Rational(abs(self.numer), abs(self.denom))

    # ---- Comparisons ----
    def __eq__(self, rhs):
        # After reduction and sign normalization, structural equality works
        return self.numer == rhs.numer and self.denom == rhs.denom

    def __lt__(self, rhs):
        # Cross-multiply to compare without converting to float
        return self.numer * rhs.denom < rhs.numer * self.denom

    def __le__(self, rhs):
        return self == rhs or self < rhs

    def __gt__(self, rhs):
        return not self <= rhs

    def __ge__(self, rhs):
        return not self < rhs

    # ---- Utilities ----
    @staticmethod
    def gcd(a, b):
        # Euclidean algorithm with absolute values
        a, b = abs(a), abs(b)
        while b != 0:
            a, b = b, a % b
        return a if a != 0 else 1  # avoid returning 0 to keep reduce() safe

    def fix_sign(self):  # takes '-' sign out of the denominator
        if self.denom < 0:  # ensure denominator is non-negative
            self.numer = -self.numer
            self.denom = -self.denom

    def reduce(self):  # reduce the fraction
        if self.numer == 0:
            # canonical zero form: 0/1
            self.denom = 1
            return
        d = self.gcd(self.numer, self.denom)
        if d > 1:
            self.numer //= d
            self.denom //= d
        self.fix_sign()

    # ---- Representations ----
    def __str__(self):
        return f"{self.numer}/{self.denom}"

    def __repr__(self):
        return f"Rational({self.numer}, {self.denom})"

    # ---- Conversions ----
    def to_float(self):
        return self.numer / self.denom

    @classmethod
    def from_int(cls, n: int):
        return cls(n, 1)


# ---- Quick demo / tests ----
if __name__ == "__main__":
    a = Rational(1, 2)
    b = Rational(3, 4)

    print("a =", a)                  # 1/2
    print("b =", b)                  # 3/4
    print("a + b =", a + b)          # 5/4
    print("a - b =", a - b)          # -1/4
    print("a * b =", a * b)          # 3/8
    print("a / b =", a / b)          # 2/3
    print("float(a) =", a.to_float())# 0.5
    print("a == Rational(2,4)?", a == Rational(2, 4))  # True (reduced)
    print("Rational(-2, -4) ->", Rational(-2, -4))     # 1/2 (sign fixed + reduced)
    print("Rational(0, 5) ->", Rational(0, 5))         # 0/1
```

```python

import math
from numbers import Real

class Radian:
    """Angle value stored in radians, with operator overloading for +, -, *, /."""
    __slots__ = ("value",)

    def __init__(self, value: Real):
        self.value = float(value)

    # ---------- Representations ----------
    def __repr__(self):
        return f"Radian({self.value})"

    def __str__(self):
        # human-friendly: show both radians and degrees
        return f"{self.value} rad ({math.degrees(self.value):.2f}°)"

    # ---------- Conversions ----------
    def to_degrees(self):
        return math.degrees(self.value)

    # ---------- Helpers ----------
    @staticmethod
    def _coerce_other_as_radians(other):
        """
        Allow operations with another Radian or a plain number.
        Numbers are treated as radians.
        """
        if isinstance(other, Radian):
            return other.value
        if isinstance(other, Real):
            return float(other)
        return NotImplemented

    # ---------- Addition ----------
    def __add__(self, other):
        other_val = self._coerce_other_as_radians(other)
        if other_val is NotImplemented:
            return NotImplemented
        return Radian(self.value + other_val)

    def __radd__(self, other):
        # addition is commutative for these types
        return self.__add__(other)

    # ---------- Subtraction ----------
    def __sub__(self, other):
        other_val = self._coerce_other_as_radians(other)
        if other_val is NotImplemented:
            return NotImplemented
        return Radian(self.value - other_val)

    def __rsub__(self, other):
        # (number or Radian) - Radian
        other_val = self._coerce_other_as_radians(other)
        if other_val is NotImplemented:
            return NotImplemented
        return Radian(other_val - self.value)

    # ---------- Multiplication (scale by a number) ----------
    def __mul__(self, factor):
        if isinstance(factor, Real):
            return Radian(self.value * float(factor))
        return NotImplemented

    def __rmul__(self, factor):
        return self.__mul__(factor)

    # ---------- True division (divide by a number) ----------
    def __truediv__(self, divisor):
        if not isinstance(divisor, Real):
            return NotImplemented
        if divisor == 0:
            raise ZeroDivisionError("Cannot divide by zero")
        return Radian(self.value / float(divisor))

    # Optional: equality for simple testing / assertions
    def __eq__(self, other):
        other_val = self._coerce_other_as_radians(other)
        if other_val is NotImplemented:
            return NotImplemented
        # tolerant comparison could be added; keep exact for simplicity
        return self.value == other_val


# --------- Example usage ----------
if __name__ == "__main__":
    r1 = Radian(math.pi)
    r2 = Radian(math.pi / 2)

    print(f"R1 in degrees: {r1.to_degrees()}")
    print(f"R2 in degrees: {r2.to_degrees()}")

    r3 = r1 + r2            # uses __add__
    print(f"R1 + R2: {r3.value}")

    r4 = r1 - r2            # uses __sub__
    print(f"R1 - R2: {r4.value}")

    r5 = r1 * 2             # uses __mul__
    print(f"R1 * 2: {r5.value}")

    r6 = r1 / 2             # uses __truediv__
    print(f"R1 / 2: {r6.value}")

    # Mixed operations with plain numbers (treated as radians):
    r7 = r1 + math.pi/4     # Radian + float
    r8 = math.pi/4 + r1     # float + Radian (uses __radd__)
    r9 = math.pi - r2       # float - Radian (uses __rsub__)
    print(f"R1 + π/4: {r7.value}")
    print(f"π/4 + R1: {r8.value}")
    print(f"π - R2: {r9.value}")

    # Pretty printing via __str__ and debugging via __repr__
    print("Pretty:", r1)
    print("Debug :", repr(r1))
```

```python
#Try yourself
class Celsius:
    def __init__(self, value):
        self.value = value

    def to_fahrenheit(self):
        return (self.value * 9/5) + 32

    def __add__(self, other):
        if isinstance(other, Celsius):
            return Celsius(self.value + other.value)
        else:
            raise TypeError("Operand must be an instance of Celsius")

    def __sub__(self, other):
        if isinstance(other, Celsius):
            return Celsius(self.value - other.value)
        else:
            raise TypeError("Operand must be an instance of Celsius")

    def __mul__(self, factor):
        return Celsius(self.value * factor)

    def __truediv__(self, divisor):
        if divisor != 0:
            return Celsius(self.value / divisor)
        else:
            raise ValueError("Cannot divide by zero")

# Example usage
c1 = Celsius(25)
c2 = Celsius(10)

print(f"C1 in Fahrenheit: {c1.to_fahrenheit()}")
print(f"C2 in Fahrenheit: {c2.to_fahrenheit()}")

c3 = c1 + c2
print(f"C1 + C2 in Celsius: {c3.value}")

c4 = c1 - c2
print(f"C1 - C2 in Celsius: {c4.value}")

c5 = c1 * 2
print(f"C1 * 2 in Celsius: {c5.value}")

c6 = c1 / 2
print(f"C1 / 2 in Celsius: {c6.value}")

```

```python

import math
 
class Circle:
 
    def __init__(self, radius):
        self.__radius = radius
 
    def setRadius(self, radius):
        self.__radius = radius
 
    def getRadius(self):
        return self.__radius
 
    def area(self):
        return math.pi * self.__radius ** 2
 
    def __add__(self, another_circle):
        return Circle( self.__radius + another_circle.__radius )
 
    def __gt__(self, another_circle):
        return self.__radius > another_circle.__radius
 
    def __lt__(self, another_circle):
        return self.__radius < another_circle.__radius
 
    def __str__(self):
        return "Circle with radius " + str(self.__radius)
 
c1 = Circle(4)
print(c1.getRadius())
 
c2 = Circle(5)
print(c2.getRadius())
 
c3 = c1 + c2
print(c3.getRadius())
 
print( c3 > c2) # Became possible because we have added __gt__ method
 
print( c1 < c2) # Became possible because we have added __lt__ method
 
print(c3) # Became possible because we have added __str__ method

```
