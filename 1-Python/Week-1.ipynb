```python
# 3) Write `unique_preserve_order(seq)` → remove duplicates while preserving first occurrence.

def unique_preserve_order(data):
    list_result = []
    for i in data:
        if i not in list_result:
            list_result.append(i)
    return list_result
        

data = [1, 8, 9, 1, 4, 3, 5, 4, 6, 9, 2, 1]
result = unique_preserve_order(data)
print(result)

# Alernate Solution
'''
def unique_preserve_order(seq):
    return list(dict.fromkeys(seq))
'''
```

```python
'''
4) Write a program that computes the perimeter and area of a triangle.
Your program should accept one of the following input modes:
 
By Sides: three side lengths a, b, c
Right Triangle by Legs: two legs a, b (hypotenuse derived)
Equilateral: side s
Isosceles: equal side e and base b
 
The program must:
 
Validate that the inputs form a valid triangle
Print the perimeter and area.
 
 
Input Format
 
First line: a mode keyword from: SIDES, RIGHT, EQUILATERAL, ISOSCELES, COORDS
Following lines (depending on mode):
 
SIDES: three floats a b c
RIGHT: two floats a b
EQUILATERAL: one float s
ISOSCELES: two floats e b
COORDS: six floats x1 y1 x2 y2 x3 y3
 
 
Geometry Formulas to Use
1) Validity (Triangle Inequality)
For sides a, b, c (all > 0):
 
Valid if a + b > c, a + c > b, b + c > a.
 
 
2) Perimeter
 
By sides: P = a + b + c
Right triangle by legs: c = √(a² + b²), then P = a + b + c
Equilateral with side s: P = 3s
Isosceles with equal sides e and base b: P = 2e + b
By coordinates: compute side lengths using distance formula and sum.
 
Distance between (x1, y1) and (x2, y2):
d = √((x2 - x1)² + (y2 - y1)²)
 
3) Area
 
Heron’s Formula (any triangle with sides a, b, c):
 
s = (a + b + c) / 2 (semiperimeter)
Area = √(s (s - a) (s - b) (s - c))
 
 
Right Triangle with legs a, b:
 
Area = (a * b) / 2
 
 
Equilateral with side s:
 
Area = (√3 / 4) * s²
 
 
Isosceles with equal sides e and base b:
 
Height h = √(e² - (b/2)²)
Area = (b * h) / 2
 
 
By Coordinates (Shoelace formula for triangle):
 
Area = |x1(y2 − y3) + x2(y3 − y1) + x3(y1 − y2)| / 2
'''
```

```python
import math

def calculate_triangle():
    mode = input("Enter mode (SIDES, RIGHT, EQUILATERAL, ISOSCELES, COORDS): ").strip().upper()
    
    a, b, c = 0.0, 0.0, 0.0
    valid = False

    try:
        if mode == "SIDES":
            a, b, c = map(float, input("Enter sides a b c: ").split())
        
        elif mode == "RIGHT":
            a, b = map(float, input("Enter legs a b: ").split())
            c = math.sqrt(a**2 + b**2)
        
        elif mode == "EQUILATERAL":
            s = float(input("Enter side s: "))
            a = b = c = s
        
        elif mode == "ISOSCELES":
            e, b_side = map(float, input("Enter equal side e and base b: ").split())
            a, b, c = e, e, b_side
            
        elif mode == "COORDS":
            x1, y1, x2, y2, x3, y3 = map(float, input("Enter x1 y1 x2 y2 x3 y3: ").split())
            # Distance formula: d = sqrt((x2-x1)^2 + (y2-y1)^2)
            a = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)
            b = math.sqrt((x3 - x2)**2 + (y3 - y2)**2)
            c = math.sqrt((x1 - x3)**2 + (y1 - y3)**2)
        
        else:
            print("Invalid mode selected.")
            return

        # 1) Validity Check (Triangle Inequality)
        if a + b > c and a + c > b and b + c > a and all(side > 0 for side in [a, b, c]):
            valid = True
        else:
            print("Error: The provided dimensions do not form a valid triangle.")
            return

        # 2) Perimeter Calculation
        perimeter = a + b + c

        # 3) Area Calculation
        # We use Heron's Formula as a universal fallback, or specific formulas for efficiency
        if mode == "RIGHT":
            area = (a * b) / 2
        elif mode == "EQUILATERAL":
            area = (math.sqrt(3) / 4) * (a**2)
        elif mode == "ISOSCELES":
            height = math.sqrt(a**2 - (c / 2)**2)
            area = (c * height) / 2
        else:
            # Heron's Formula
            s_peri = perimeter / 2
            area = math.sqrt(s_peri * (s_peri - a) * (s_peri - b) * (s_peri - c))

        print(f"\n--- Results ---")
        print(f"Sides: {a:.2f}, {b:.2f}, {c:.2f}")
        print(f"Perimeter: {perimeter:.2f}")
        print(f"Area: {area:.2f}")

    except ValueError:
        print("Invalid input. Please enter numeric values.")

calculate_triangle()
```

```python
session_data = {
    "user_id": "  u_1045 ",
    "device": "mobile",
    "watch_history": [
        ("VID101", "the.godfather  ", "Crime", 175),
        ("VID102", "inception-HD", "Sci-Fi", 148),
        ("VID103", " The.Dark.Knight ", "Action", 152),
        ("VID104", "inception-HD", "Sci-Fi", 30)
    ]
}

def clean_user_id(uid):
    return uid.strip().upper()

def clean_title(raw):
    s = raw.strip().replace('.', ' ').replace('-', ' ')
    s = ' '.join(s.split())  
    return s.title()


cleaned_user_id = clean_user_id(session_data["user_id"])  

# Cleaned titles list
cleaned_titles = []
for rec in session_data["watch_history"]:
    raw_title = rec[1]
    cleaned_titles.append(clean_title(raw_title))


unique_genres = []
for rec in session_data["watch_history"]:
    genre = rec[2]
    if genre not in unique_genres:
        unique_genres.append(genre)


accessed_fields = []
for rec in session_data["watch_history"]:
    
    video_id, raw_title, genre, watch_time = rec
    accessed_fields.append((genre, watch_time))


watch_time_by_genre = {}
for rec in session_data["watch_history"]:
    genre = rec[2]
    minutes = rec[3]
    if genre in watch_time_by_genre:
        watch_time_by_genre[genre] += minutes
    else:
        watch_time_by_genre[genre] = minutes


most_watched_genre = max(watch_time_by_genre, key=watch_time_by_genre.get)
summary = f"USER {cleaned_user_id} shows highest engagement with {most_watched_genre.upper()} content"


print("Cleaned user_id:", cleaned_user_id)
print("Cleaned video titles:", cleaned_titles)
print("Unique genres:", unique_genres)
print("(genre, watch_time) accessed from each tuple:", accessed_fields)
print("Watch time by genre:", watch_time_by_genre)
print("Data Science Insight:", summary)


total_minutes = sum([rec[3] for rec in session_data["watch_history"]])
recomputed_total = sum(watch_time_by_genre.values())
print("Total minutes (raw):", total_minutes)
print("Total minutes (aggregated):", recomputed_total)
```
