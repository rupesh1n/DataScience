and non-alphanumeric characters). Example: is_palindrome("A man, a plan, a canal: Panama") → True.

```python
def is_palindrome(s):
	# will do operation on string s to make lower case and removing extra space
	new_string = s.lower().strip().replace(",", "").replace(":", "").replace(" ", "")
	#new_string = "".join(char for char in s if char.isalpha()).lower()
	
	if new_string == new_string[::-1]:
		return True
	return False

print(is_palindrome("A man, a plan, a canal: Panama")) 
```

Challenge: Write a function most_frequent_char(s) that returns the most frequent character in a string
s. If multiple characters have the same frequency, return the one that appears first alphabetically.
Example: most_frequent_char("banana") → 'a'.

```python
def most_frequent_char(s):
    if not s:
        return None
    
    frequency = {}
    for char in s:
        frequency[char] = frequency.get(char, 0) + 1
        
    max_freq = max(frequency.values())
    candidates = [char for char, count in frequency.items() if count == max_freq]
    return min(candidates)

print(most_frequent_char("banana"))
print(most_frequent_char("aabbcc"))
```

Basic: Write a function remove_duplicates(lst) that takes a list lst and returns a new list with duplicates removed, preserving the original order. Example: remove_duplicates([1, 2, 2, 3, 1]) → [1, 2, 3].

```python
li = [1, 2, 2, 3, 1]
def remove_duplicates(li):
    result = []
    visited = set()

    for i in li:
        if i not in visited:
            result.append(i)
            visited.add(i)
    return result

print("Unique element list is : ", remove_duplicates(li))
```

Challenge: Write a function rotate_list(lst, k) that rotates a list lst to the right by k positions. Example:
rotate_list([1, 2, 3, 4, 5], 2) → [4, 5, 1, 2, 3].

```python
def rotate_list(lst, k):
    if not lst:
        return []
    length = len(lst)

    min_rotation = k % length

    if min_rotation == 0:
        return lst
    
    # Slice the list and concatenate to perform rotation
    rotated_list = lst[-min_rotation:] + lst[:-min_rotation]
    return rotated_list
    
li = [1, 2, 3, 4, 5]
print("Original List : ", li )
print("Roated List : ", rotate_list(li, 2) )
```

4. Dictionaries
Basic: Write a function word_count(s) that takes a string s and returns a dictionary with the count of
each word. Example: word_count("the quick brown fox") → {'the': 1, 'quick': 1, 'brown': 1, 'fox': 1}.

```python
def word_count(s: str) -> dict:
    words = s.split()
    counts = {}
    
    for word in words:
        word = word.lower() #// Making Key unique 
        if word in counts:
            counts[word] += 1
        else:
            counts[word] = 1
            
    return counts

sentence1 = "Python is fun and python is powerful"
counts1 = word_count(sentence1)
print(f"'{sentence1}' -> {counts1}")
```

Intermediate: Write a function merge_dicts(dict1, dict2) that merges two dictionaries, summing values
for common keys. Example: merge_dicts({'a': 1, 'b': 2}, {'b': 3, 'c': 4}) → {'a': 1, 'b': 5, 'c': 4}.

```python
def merge_dicts(dict1, dict2):
    merged_dict = dict1.copy()  # Start with a copy of the first dictionary
    for k, v in dict2.items():
        if k in merged_dict:
            # If the key exists in both dictionaries, sum the values
            merged_dict[k] += v
        else:
            # Otherwise, add the new key-value pair
            merged_dict[k] = v
    return merged_dict

d1 = {'a': 1, 'b': 2}
d2 = {'b': 3, 'c': 4}
merged = merge_dicts(d1, d2)
print(merged)
```

Challenge: Write a function group_by_first_letter(words) that takes a list of words and returns a
dictionary where keys are the first letters of words, and values are lists of words starting with that letter.
Example: group_by_first_letter(['apple', 'ant', 'banana']) → {'a': ['apple', 'ant'], 'b': ['banana']}.

```python
def group_by_first_letter(words):
    grouped_dict = {}
    
    for word in words:
        if not word:
            continue
            
        first_letter = word[0]
        
        # If the letter is not yet a key, create it with an empty list
        if first_letter not in grouped_dict:
            grouped_dict[first_letter] = []
        
        # Append the word to the corresponding list
        grouped_dict[first_letter].append(word)
        
    return grouped_dict

word_list = ['apple', 'ant', 'banana', 'cat', 'dog', 'bear']
result = group_by_first_letter(word_list)
print(result)
# Output: {'a': ['apple', 'ant'], 'b': ['banana', 'bear'], 'c': ['cat'], 'd': ['dog']}
```

Write a function analyze_text(s) that takes a string s and returns a dictionary with:
The total number of unique words (use a set).
The most frequent word (use a dictionary).
The reversed string (use string operations).
A sorted list of word lengths (use a list).

```python
def analyze_text(s):
    # Normalize input: split into lowercase words while ignoring common punctuation
    words = s.lower().replace('.', '').replace(',', '').split()
    
    # 1. Total number of unique words using a set
    unique_count = len(set(words))
    
    # 2. Most frequent word using a dictionary manually
    counts = {}
    for word in words:
        counts[word] = counts.get(word, 0) + 1
    
    # Determine the most frequent word manually
    most_frequent = None
    if counts:
        most_frequent = max(counts, key=counts.get)
    
    # 3. Reversed string using string slicing
    reversed_s = s[::-1]
    
    # 4. Sorted list of word lengths
    word_lengths = sorted([len(word) for word in words])
    
    return {
        "unique_words": unique_count,
        "most_frequent": most_frequent,
        "reversed_string": reversed_s,
        "word_lengths": word_lengths
    }


print(analyze_text("I have a mug. It is blue. Jub is also blue"))
```

Write a Python program to compute the square of the first N Fibonacci numbers using
the map function and generate a list of the numbers.

```python
def fibonacci(N):
    fib_sequence = []
    a, b = 0, 1
    for _ in range(N):
        fib_sequence.append(a)
        a, b = b, a + b
    return fib_sequence

def square_list(numbers):
    squared_numbers_map = map(lambda x: x * x, numbers)
    return list(squared_numbers_map)

fib_numbers = fibonacci(10)
print(f"The first 10 fibonacci numbers are: {fib_numbers}")

print(f"The squares of these numbers are: {square_list(fib_numbers)}")
```

3. Filter Functions:
Write a Python program to filter out prime numbers from a list of integers using the filter
function.

```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

numbers = list(range(1, 21))

prime_numbers_list = list(filter(is_prime, numbers))

print(f"Original list: {numbers}")
print(f"Prime numbers: {prime_numbers_list}")
```

4. Reduce Functions:
Write a Python program to concatenate a list of strings using the reduce function. For
example, given a list ['Hello', 'world', 'from', 'Python'], concatenate the strings to form a
single sentence.

```python
from functools import reduce

def concatenate_strings(li):

    sentence = reduce(lambda x, y: x + " " + y, li)
    return sentence

words = ['Hello', 'world', 'from', 'Python']
result = concatenate_strings(words)

print(f"Original list: {words}")
print(f"Concatenated sentence: {result}")
```

Write a Python program to calculate the factorial of a number using the reduce function.
For example, given a number n, calculate the factorial of n.

```python
from functools import reduce

def factorial_reduce_lambda(n):
    if not isinstance(n, int) or n < 0:
        raise ValueError("n must be a non-negative integer")
        
    if n == 0 or n == 1:
        return 1
    else:
        result = reduce(lambda x, y: x * y, range(1, n + 1))
        return result

number = 5
print(f"The factorial of {number} is: {factorial_reduce_lambda(number)}")
```
