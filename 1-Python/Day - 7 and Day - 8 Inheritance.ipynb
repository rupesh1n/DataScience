```python
'''
Shallow Copy (copy.copy):

Creates a new Car object.
But the engine inside it is still a reference to the sameEngine object.
So, changes to engine.horsepower affect both original_car and shallow_copied_car.

Deep Copy (copy.deepcopy):
Creates a new Car object and a new Engine object.
So, changes to engine.horsepower in the original do not affect the deep copy.
'''

import copy

class Engine:
    def __init__(self, horsepower):
        self.horsepower = horsepower

    def __repr__(self):
        return f"Engine({self.horsepower} HP)"

class Car:
    def __init__(self, brand, engine):
        self.brand = brand
        self.engine = engine

    def __repr__(self):
        return f"Car(brand={self.brand}, engine={self.engine})"

# Original object
original_engine = Engine(150)
original_car = Car("Toyota", original_engine)

# Shallow copy
shallow_copied_car = copy.copy(original_car)

# Deep copy
deep_copied_car = copy.deepcopy(original_car)

# Modify the engine horsepower
original_car.engine.horsepower = 200

# Output results
print("Original Car:", original_car)
print("Shallow Copied Car:", shallow_copied_car)
print("Deep Copied Car:", deep_copied_car)
```

#### Inheritence and Operator overloading

```python

class Temperature:
    def __init__(self, value):
        self.value = value

    def to_fahrenheit(self):
        raise NotImplementedError("This method should be implemented by subclasses")

    def to_celsius(self):
        raise NotImplementedError("This method should be implemented by subclasses")

class Celsius(Temperature):
    def to_fahrenheit(self):
        return (self.value * 9/5) + 32

    def __add__(self, other):
        if isinstance(other, Celsius):
            return Celsius(self.value + other.value)
        else:
            raise TypeError("Operand must be an instance of Celsius")

    def __sub__(self, other):
        if isinstance(other, Celsius):
            return Celsius(self.value - other.value)
        else:
            raise TypeError("Operand must be an instance of Celsius")

    def __mul__(self, factor):
        return Celsius(self.value * factor)

    def __truediv__(self, divisor):
        if divisor != 0:
            return Celsius(self.value / divisor)
        else:
            raise ValueError("Cannot divide by zero")

class Fahrenheit(Temperature):
    def to_celsius(self):
        return (self.value - 32) * 5/9

    def __add__(self, other):
        if isinstance(other, Fahrenheit):
            return Fahrenheit(self.value + other.value)
        else:
            raise TypeError("Operand must be an instance of Fahrenheit")

    def __sub__(self, other):
        if isinstance(other, Fahrenheit):
            return Fahrenheit(self.value - other.value)
        else:
            raise TypeError("Operand must be an instance of Fahrenheit")

    def __mul__(self, factor):
        return Fahrenheit(self.value * factor)

    def __truediv__(self, divisor):
        if divisor != 0:
            return Fahrenheit(self.value / divisor)
        else:
            raise ValueError("Cannot divide by zero")

# Example usage
c1 = Celsius(25)
c2 = Celsius(10)

print(f"C1 in Fahrenheit: {c1.to_fahrenheit()}")
print(f"C2 in Fahrenheit: {c2.to_fahrenheit()}")

c3 = c1 + c2
print(f"C1 + C2 in Celsius: {c3.value}")

c4 = c1 - c2
print(f"C1 - C2 in Celsius: {c4.value}")

c5 = c1 * 2
print(f"C1 * 2 in Celsius: {c5.value}")

c6 = c1 / 2
print(f"C1 / 2 in Celsius: {c6.value}")

f1 = Fahrenheit(77)
f2 = Fahrenheit(50)

print(f"F1 in Celsius: {f1.to_celsius()}")
print(f"F2 in Celsius: {f2.to_celsius()}")

f3 = f1 + f2
print(f"F1 + F2 in Fahrenheit: {f3.value}")

f4 = f1 - f2
print(f"F1 - F2 in Fahrenheit: {f4.value}")

f5 = f1 * 2
print(f"F1 * 2 in Fahrenheit: {f5.value}")

f6 = f1 / 2
print(f"F1 / 2 in Fahrenheit: {f6.value}")

```

```python
import math
 
class Shape(object):
 
    def __init__(self, color='black', filled=False):
        self.__color = color
        self.__filled = filled
 
    def get_color(self):
        return self.__color
 
    def set_color(self, color):
        self.__color = color
 
    def get_filled(self):
        return self.__filled
 
    def set_filled(self, filled):
        self.__filled = filled
 
 
class Rectangle(Shape):
 
    def __init__(self, length, breadth):
        super(Rectangle,self).__init__()
        self.__length = length
        self.__breadth = breadth
 
    def get_length(self):
        return self.__length
 
    def set_length(self, length):
        self.__length = length
 
    def get_breadth(self):
        return self.__breadth
 
    def set_breadth(self, breadth):
        self.__breadth = breadth
 
    def get_area(self):
        return self.__length * self.__breadth
 
    def get_perimeter(self):
        return 2 * (self.__length + self.__breadth)
 
 
class Circle(Shape):
    def __init__(self, radius):
        super(Circle,self).__init__()
        self.__radius = radius
 
    def get_radius(self):
        return self.__radius
 
    def set_radius(self, radius):
        self.__radius = radius
 
    def get_area(self):
        return math.pi * self.__radius ** 2
 
    def get_perimeter(self):
        return 2 * math.pi * self.__radius
 

def fnArea(ref):
    print(ref.get_area())

def fnPerimeter(ref):
    print(ref.get_perimeter())


r1 = Rectangle(10.5, 2.5)
 
print("Area of rectangle r1:", r1.get_area())
print("Perimeter of rectangle r1:", r1.get_perimeter())
print("Color of rectangle r1:", r1.get_color())
print("Is rectangle r1 filled ? ", r1.get_filled())
r1.set_filled(True)
print("Is rectangle r1 filled ? ", r1.get_filled())
r1.set_color("orange")
print("Color of rectangle r1:", r1.get_color())
 
c1 = Circle(12)
 
print("\nArea of circle c1:", format(c1.get_area(), "0.2f"))
print("Perimeter of circle c1:", format(c1.get_perimeter(), "0.2f"))
print("Color of circle c1:", c1.get_color())
print("Is circle c1 filled ? ", c1.get_filled())
c1.set_filled(True)
print("Is circle c1 filled ? ", c1.get_filled())
c1.set_color("blue")
print("Color of circle c1:", c1.get_color())

fnArea(c1)
fnPerimeter(c1)

fnArea(r1)
fnPerimeter(r1)
```

### Abstract Classes and Interfaces 
### collections.abc module
Abstract Base Classes (ABCs) are classes that cannot be instantiated directly and are designed to be subclassed. They serve as blueprints or contracts that define what methods a subclass must implement.
Key Benefits:
	• Enforce consistent interfaces across related classes
	• Provide partial implementation while requiring specific methods
	• Enable better code organization and maintainability
	• Support polymorphism with guaranteed method availability

```python
'''
Python's abc module provides the infrastructure for defining Abstract Base Classes.
from abc import ABC, abstractmethod
'''
# Basic example
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
    
    @abstractmethod
    def perimeter(self):
        pass
```

```python
# importing abstract base classes module
import abc

class GetterSetter(abc.ABC):
    '''
    ABSTRACT BASE CLASSES:

    -  An abstract base class is a kind of 'model' for other classes to be defined.
        - It is not designed to construct instances, but can be subclassed by regular classes

    - Abstract classes can define interface, or methods that must be implemented by its subclasses.

    '''

    # Abstract classes are not designed to be instantiated, only to be subclassed

    #  decorator for abstract class
    @abc.abstractmethod
    def set_val(self, input):
        """set the value in the instance"""
        return

    @abc.abstractmethod
    def get_val(self):
        """Get and return a value from the instance..."""
        return

# Inheriting from the above abstract class
class MyClass(GetterSetter):

    # methods overriding in the GetterSetter
    def set_val(self, input):
        self.val = input

    def get_val(self):
        return self.val
# Instantiate
x = MyClass()
print(x) # prints the instance <__main__.MyClass object at 0x10218ee48>
x.set_val("Hello")
print(x.get_val())

#x = GetterSetter() #throws error, abstract classes can't be instantiated
```

```python
#Creating Abstract Classes - Comprehensive Example

from abc import ABC, abstractmethod
from typing import List
class Employee(ABC):
    """Abstract base class for all employees"""
    
    def __init__(self, name: str, employee_id: str):
        self.name = name
        self.employee_id = employee_id
    
    @abstractmethod
    def calculate_salary(self) -> float:
        """Calculate employee's salary - must be implemented"""
        pass
    
    @abstractmethod
    def get_benefits(self) -> List[str]:
        """Get list of employee benefits"""
        pass
    
    # Concrete method (shared implementation)
    def display_info(self):
        return f"Employee: {self.name} (ID: {self.employee_id})"
    
    @property
    @abstractmethod
    def employment_type(self) -> str:
        """Type of employment"""
        pass
# Abstract class with partial implementation
class SalariedEmployee(Employee):
    """Abstract class for salaried employees"""
    
    def __init__(self, name: str, employee_id: str, annual_salary: float):
        super().__init__(name, employee_id)
        self.annual_salary = annual_salary
    
    def calculate_salary(self) -> float:
        return self.annual_salary / 12  # Monthly salary
    
    def get_benefits(self) -> List[str]:
        return ["Health Insurance", "Retirement Plan", "Paid Time Off"]
    
    @property
    def employment_type(self) -> str:
        return "Salaried"
    
    @abstractmethod
    def get_department(self) -> str:
        """Department-specific method"""
        pass
```

```python
# Concrete implementations
class FullTimeEmployee(SalariedEmployee):
    def __init__(self, name: str, employee_id: str, annual_salary: float, department: str):
        super().__init__(name, employee_id, annual_salary)
        self.department = department
    
    def get_department(self) -> str:
        return self.department
    
    def get_benefits(self) -> List[str]:
        benefits = super().get_benefits()
        benefits.extend(["Stock Options", "Professional Development"])
        return benefits
        
class ContractEmployee(Employee):
    def __init__(self, name: str, employee_id: str, hourly_rate: float, hours_per_month: int):
        super().__init__(name, employee_id)
        self.hourly_rate = hourly_rate
        self.hours_per_month = hours_per_month
    
    def calculate_salary(self) -> float:
        return self.hourly_rate * self.hours_per_month
    
    def get_benefits(self) -> List[str]:
        return ["Basic Health Coverage"]
    
    @property
    def employment_type(self) -> str:
        return "Contract"
# Usage example
def process_employee(employee: Employee):
    print(employee.display_info())
    print(f"Monthly Salary: ${employee.calculate_salary():.2f}")
    print(f"Benefits: {', '.join(employee.get_benefits())}")
    print(f"Type: {employee.employment_type}\n")
    
# Create instances
ft_employee = FullTimeEmployee("Alice Johnson", "FT001", 75000, "Engineering")
contract_employee = ContractEmployee("Bob Smith", "CT001", 50, 160)
process_employee(ft_employee)
process_employee(contract_employee)
```
