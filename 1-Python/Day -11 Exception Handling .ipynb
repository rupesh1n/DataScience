5.1 Syntax Errors vs Exceptions
Syntax Errors
	• Occur when the code violates Python’s grammar rules.
	• Detected before the program runs.

Common Syntax Errors:
1.	1. Missing colons:
if x > 5  # SyntaxError

2.	2. Incorrect indentation:
def greet():
print("Hi")  # IndentationError

3.	3. Unmatched brackets:
mylist = [1, 2, 3  # SyntaxError

4.	4. Invalid variable names:
2var = 10  # SyntaxError
    

Runtime Exceptions
Exceptions occur during program execution even when syntax is correct. They represent logical errors or unexpected conditions.
Common Runtime Exceptions:

1. ZeroDivisionError - Division by zero
   result = 10 / 0

2. ValueError - Invalid value for operation
   num = int('abc')

3. TypeError - Wrong type for operation
   result = '2' + 2

4. IndexError - Index out of range
   mylist = [1, 2, 3]
   print(mylist[10])

5. KeyError - Key not in dictionary
   mydict = {'name': 'John'}
   print(mydict['age'])

6. AttributeError - Invalid attribute access
   text = 'hello'
   text.append('!')

7. FileNotFoundError - File doesn't exist
   f = open('nonexistent.txt')

8. ImportError / ModuleNotFoundError - Module import failure
   import nonexistent_module


```python
#try…except Block

try:
    x = int(input("Enter a number: "))
except ValueError:
    print("Invalid input!")
```

```python
#Example - Division with Error Handling:
def safe_divide(a, b):
    try:
        result = a / b
        return result
    except ZeroDivisionError:
        print('Error: Cannot divide by zero!')
        return None

print(safe_divide(10, 2))   # Output: 5.0
print(safe_divide(10, 0))   # Output: Error: Cannot divide by zero! None
```

```python
#Catching Multiple Exceptions
#Method 1: Separate except blocks
try:
    value = int(input('Enter number: '))
    result = 100 / value
except ValueError:
    print('Invalid input!')
except ZeroDivisionError:
    print('Cannot divide by zero!')
```

```python
#Method 2: Tuple of exceptions
try:
    value = int(input('Enter number: '))
    result = 100 / value
except (ValueError, TypeError, ZeroDivisionError) as e:
    print(f'An error occurred: {e}')
```

```python
#5.4 else and finally Clauses
#else clause - Runs only if no exception occurred:
try:
    file = open('data.txt', 'r')
except FileNotFoundError:
    print('File not found!')
else:
    # Only runs if file opened successfully
    data = file.read()
    file.close()
```

```python
#5.5 finally clause - Always executes, regardless of exceptions:
file = None
try:
    file = open('data.txt', 'r')
    data = file.read()
except FileNotFoundError:
    print('File not found')
finally:
    # Always executes
    if file:
        file.close()
```

```python
# =============================================================================
# 4. FINALLY BLOCK - ALWAYS EXECUTES
# =============================================================================
print("4. Finally Block Example:")
def read_file_demo(filename):
    file_handle = None
    try:
        print(f"Attempting to open {filename}")
        file_handle = open(filename, 'r')
        content = file_handle.read()
        print(f"File content: {content}")
    except FileNotFoundError:
        print(f"Error: File {filename} not found!")
    except PermissionError:
        print(f"Error: Permission denied for {filename}")
    finally:
        print("Finally block executing...")
        if file_handle:
            file_handle.close()
            print("File closed successfully")
        else:
            print("No file to close")
    print("Function completed\n")

read_file_demo("nonexistent.txt")
```

```python
# =============================================================================
# 7. NESTED EXCEPTION HANDLING
# =============================================================================
print("7. Nested Exception Handling:")
def nested_operations():
    try:
        print("Outer try block started")
        try:
            print("Inner try block started")
            # Simulate nested operations
            data = input("Enter a number (or press Enter for demo): ").strip()
            if not data:
                data = "10"  # Default for demo
            
            number = int(data)
            try:
                result = 100 / number
                print(f"Result: {result}")
            except ZeroDivisionError:
                print("Inner exception: Division by zero!")
                raise ValueError("Nested error: Invalid calculation")
        except ValueError as ve:
            print(f"Middle exception caught: {ve}")
            # Re-raise to outer block
            raise RuntimeError("Outer error: Processing failed") from ve
    except RuntimeError as re:
        print(f"Outer exception caught: {re}")
        print(f"Original cause: {re.__cause__}")
    finally:
        print("Nested operation cleanup completed")

nested_operations()
print()
```

5.4 User-Defined Exceptions
Custom exception classes make code more readable and allow application-specific error handling.
Creating Custom Exception Classes
Basic custom exception:



```python
class ValidationError(Exception):
    """Raised when data validation fails"""
    pass

def validate_email(email):
    if '@' not in email:
        raise ValidationError('Email must contain @')
```

```python
#Advanced custom exception with attributes:
class InsufficientFundsError(Exception):
    """Raised when withdrawal exceeds balance"""
    
    def __init__(self, balance, amount):
        self.balance = balance
        self.amount = amount
        self.shortage = amount - balance
        
        message = (
            f'Insufficient funds: balance=${balance:.2f}, '
            f'attempted=${amount:.2f}, shortage=${self.shortage:.2f}'
        )
        super().__init__(message)

# Usage
try:
    if amount > balance:
        raise InsufficientFundsError(balance, amount)
except InsufficientFundsError as e:
    print(f'Error: {e}')
    print(f'You need ${e.shortage:.2f} more')
```

5.5 Context Managers
Context managers provide a clean way to manage resources, ensuring proper acquisition and release even if exceptions occur.
The with Statement
Without context manager (manual cleanup):


```python
#With context manager (automatic cleanup):
with open('data.txt', 'r') as file:
    data = file.read()
# File automatically closed, even if exception occurs
```

```python
#Creating Context Managers
#Class-based with __enter__ and __exit__:
class FileManager:
    def __init__(self, filename, mode):
        self.filename = filename
        self.mode = mode
        self.file = None
    
    def __enter__(self):
        """Called when entering with block"""
        self.file = open(self.filename, self.mode)
        return self.file
    
    def __exit__(self, exc_type, exc_val, exc_tb):
        """Called when exiting with block"""
        if self.file:
            self.file.close()
        return False  # Don't suppress exceptions

# Usage
with FileManager('test.txt', 'w') as f:
    f.write('Hello, World!')
```
