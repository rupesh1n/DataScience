```python
def celsius_to_fahrenheit(celsius):
    """
    Convert Celsius to Fahrenheit.

    Parameters:
    celsius (float): Temperature in Celsius.

    Returns:
    float: Temperature in Fahrenheit.
    """
    fahrenheit = (celsius * 9/5) + 32
    return fahrenheit

# Example usage:
celsius_temp =float(input('Enter a celsius'))
fahrenheit_temp = celsius_to_fahrenheit(celsius_temp)
print(f"{celsius_temp}°C is equal to {fahrenheit_temp}°F")
```

```python
def prime_numbers(lower_limit, upper_limit):
    primes = []
    for num in range(lower_limit, upper_limit + 1):
        if num > 1:  # Prime numbers are greater than 1
            for i in range(2, int(num ** 0.5) + 1):
                if num % i == 0:
                    break
            else:
                primes.append(num)
    return primes

x=prime_numbers(2,10)
x
```

```python
# ---------------------------------------------------------------------------
# MULTIPLE RETURN VALUES
# ---------------------------------------------------------------------------

def get_statistics(numbers):
    """
    MULTIPLE RETURN VALUES:
    -----------------------
    Python functions can return multiple values using tuples.
    Values are separated by commas and automatically packed into a tuple.
    """
    total = sum(numbers)
    average = total / len(numbers)
    maximum = max(numbers)
    minimum = min(numbers)
    
    return total, average, maximum, minimum

# Unpack returned values
data = [10, 20, 30, 40, 50]
sum_val, avg_val, max_val, min_val = get_statistics(data)

print(f"Sum: {sum_val}")
print(f"Average: {avg_val}")
print(f"Maximum: {max_val}")
print(f"Minimum: {min_val}")
```

```python
# =============================================================================
# 4.2 VARIABLE SCOPE AND NAMESPACE
# =============================================================================

WHAT IS SCOPE?
--------------
Scope determines the visibility and lifetime of variables.
Python follows the LEGB rule for variable lookup.

LEGB RULE:
----------
L - Local: Variables defined inside a function
E - Enclosing: Variables in the local scope of enclosing functions
G - Global: Variables defined at the module level
B - Built-in: Python's built-in names (print, len, etc.)

Python searches for variables in this order: L → E → G → B
```

```python
"""
'global' KEYWORD:
-----------------
- Used to modify global variables from inside a function
- Without 'global', assignment creates a new local variable
- Use sparingly as it can make code harder to understand
"""
counter = 0  # Global variable

def increment_counter():
    """Modify global variable using global keyword"""
    global counter
    counter += 1
    print(f"Counter inside function: {counter}")

print(f"Counter before: {counter}")
increment_counter()
increment_counter()
print(f"Counter after: {counter}")
```

```python
"""
ENCLOSING SCOPE:
----------------
Applies to nested functions. Inner function can access variables
from outer (enclosing) function.
"""

def outer_function():
    """Outer function with nested function"""
    outer_var = "I'm in outer function"
    
    def inner_function():
        """Inner function accessing enclosing scope"""
        print(f"Inner accessing: {outer_var}")
    
    inner_function()
    print(f"Outer: {outer_var}")

outer_function()
```

```python

"""
'nonlocal' KEYWORD:
-------------------
- Used in nested functions to modify variables in enclosing scope
- Similar to 'global' but for enclosing function scope
- Does not affect global variables
"""

def counter_function():
    """Demonstrates nonlocal keyword"""
    count = 0
    
    def increment():
        nonlocal count  # Modify enclosing scope variable
        count += 1
        return count
    
    print(f"First call: {increment()}")
    print(f"Second call: {increment()}")
    print(f"Third call: {increment()}")
    print(f"Count: {count}")
counter_function()
```

```python
# ---------------------------------------------------------------------------
#  nonlocal in Multi-Level Nesting
# ---------------------------------------------------------------------------

def outer():
    """Multi-level nested functions"""
    x = "outer"
    
    def middle():
        x = "middle"
        
        def inner():
            nonlocal x  # Refers to middle's x, not outer's
            x = "inner"
            print(f"Inner: x = {x}")
        
        inner()
        print(f"Middle: x = {x}")
    
    middle()
    print(f"Outer: x = {x}")

outer()
```

NAMESPACE:
----------
A namespace is a container that holds a mapping of names to objects.
It prevents naming conflicts and organizes code.

TYPES OF NAMESPACES:
--------------------
1. Built-in namespace: Contains built-in functions (print, len, etc.)
2. Global namespace: Module-level names
3. Local namespace: Function-level names

NAMESPACE LIFETIME:
-------------------
- Built-in: Created when Python starts, exists until interpreter quits
- Global: Created when module is loaded, exists until program ends
- Local: Created when function is called, destroyed when function returns

RECURSION:
----------
A function that calls itself to solve a problem by breaking it down
into smaller, similar sub-problems.

COMPONENTS:
-----------
1. Base Case: Condition that stops recursion (prevents infinite loop)
2. Recursive Case: Function calls itself with modified arguments

WHEN TO USE:
------------
- Tree/graph traversal
- Divide and conquer problems
- Mathematical sequences
- Problems naturally defined recursively

```python
def factorial(n):
    """
    Calculate factorial using recursion
    n! = n × (n-1) × (n-2) × ... × 1
    """
    # Base case
    if n == 0 or n == 1:
        return 1
    
    # Recursive case
    return n * factorial(n - 1)

print(f"Factorial of 5: {factorial(5)}")  # 5! = 120
print(f"Factorial of 0: {factorial(0)}")  # 0! = 1
```

```python
def fibonacci(n):
    """
    Return nth Fibonacci number
    Sequence: 0, 1, 1, 2, 3, 5, 8, 13, ...
    """
    # Base cases
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    
    # Recursive case
    return fibonacci(n - 1) + fibonacci(n - 2)

# Generate first 10 Fibonacci numbers
fib_sequence = [fibonacci(i) for i in range(10)]
print(f"First 10 Fibonacci numbers: {fib_sequence}")
```

```python
def recursive_sum(numbers):
    """Calculate sum of list using recursion"""
    # Base case: empty list
    if len(numbers) == 0:
        return 0
    
    # Recursive case: first element + sum of rest
    return numbers[0] + recursive_sum(numbers[1:])

numbers = [1, 2, 3, 4, 5]
print(f"Recursive sum of {numbers}: {recursive_sum(numbers)}")
```

```python
def countdown(n):
    """Print countdown from n to 1"""
    # Base case
    if n <= 0:
        print("Liftoff!")
        return
    
    # Recursive case
    print(n)
    countdown(n - 1)

countdown(5)
```

DEFAULT ARGUMENTS:
------------------
Parameters can have default values. If argument is not provided,
default value is used.

RULES:
------
- Default parameters must come after non-default parameters
- Default values are evaluated once when function is defined
- Mutable defaults (lists, dicts) can cause unexpected behavior

```python
def ask_ok(prompt, retries=4 , complaint='Yes or no, please!' ) : 
    while True:
        ok = input(prompt)
        if ok in ('y', 'ye', 'yes'):
            return True
        if ok in ('n', 'no', 'nop', 'nope'):
            return False
        retries = retries - 1
        if retries < 0:
            raise OSError('uncooperative user')
        print(complaint)

x=ask_ok("Do you agree?")
#x=ask_ok("Do you agree?",2,"y, ye,yes only!")
print("Hei",end=':')
```

```python
#Mutable Default Arguments (DANGER!)
def foo(num,L=[]):
    num *= 10
    L.append(num)
    return L
print(foo(1))
print(foo(6))
print(foo(77))
print(foo(31))
```

```python
#--------------------------------------------------------------
# Mutable Default Arguments (DANGER!)
# --------------------------------------------------------------

# WRONG WAY - Mutable default
def append_to_list_wrong(item, my_list=[]):
    """DON'T DO THIS! Mutable default is shared across calls"""
    my_list.append(item)
    return my_list

print(append_to_list_wrong(1))  # [1]
print(append_to_list_wrong(2))  # [1, 2] - Unexpected!
print(append_to_list_wrong(3))  # [1, 2, 3] - Same list!

# CORRECT WAY - Use None and create new list
def append_to_list_correct(item, my_list=None):
    """Correct way to handle mutable defaults"""
    if my_list is None:
        my_list = []
    my_list.append(item)
    return my_list

print(append_to_list_correct(1))  # [1]
print(append_to_list_correct(2))  # [2] - New list each time
print(append_to_list_correct(3))  # [3]
```

KEYWORD ARGUMENTS:
------------------
Arguments passed by parameter name rather than position.

BENEFITS:
---------
- More readable code
- Order doesn't matter
- Can skip optional parameters
- Self-documenting code

```python
def parrot(voltage, state='a stiff', action='voom', type='Norwegian Blue'):
    print("-- This parrot wouldn't", action,"   ")
    print("if you put", voltage, "volts through it.")
    print("-- Lovely plumage, the", type)
    print("-- It's", state, "!")

parrot(1000)                                          # 1 positional argument
parrot(voltage=1000)                                  # 1 keyword argument
parrot(voltage=1000000, action='VOOOOOM')             # 2 keyword arguments
parrot(action='VOOOOOM', voltage=1000000)             # 2 keyword arguments
parrot('a million', 'bereft of life', 'jump')         # 3 positional arguments
parrot('a thousand', state='pushing up the daisies')  # 1 positional, 1 keyword
```

```python
def book_ticket(passenger, flight, seat_class="Economy", meal="Vegetarian"):
    """Book a flight ticket"""
    print(f"Booking for: {passenger}")
    print(f"Flight: {flight}")
    print(f"Class: {seat_class}")
    print(f"Meal: {meal}")
    print("-" * 40)

# Positional arguments
book_ticket("John", "AI101")

# Keyword arguments (any order)
book_ticket(flight="BA202", passenger="Sarah", meal="Non-Veg", seat_class="Business")

# Mixed (positional first, then keyword)
book_ticket("Mike", "UA303", meal="Vegan")
```

```python
# ---------------------------------------------------------------------------
# VARIABLE-LENGTH ARGUMENTS (*args, **kwargs)
# ---------------------------------------------------------------------------

*args (ARBITRARY POSITIONAL ARGUMENTS):
----------------------------------------
- Collects extra positional arguments into a tuple
- 'args' is just convention, can be any name
- Must come after regular positional parameters

**kwargs (ARBITRARY KEYWORD ARGUMENTS):
----------------------------------------
- Collects extra keyword arguments into a dictionary
- 'kwargs' is just convention, can be any name
- Must come after *args if both are used

PARAMETER ORDER:
----------------
1. Regular positional parameters
2. *args
3. Regular keyword parameters
4. **kwargs
```

```python
# *args - Variable Positional Arguments

def sum_all(*numbers):
    """Sum any number of arguments"""
    print(f"Received {len(numbers)} numbers")
    print(f"Type: {type(numbers)}")  # tuple
    return sum(numbers)

print(sum_all(1, 2, 3))
print(sum_all(10, 20, 30, 40, 50))
print(sum_all(5))
```

```python
# ---------------------------------------------------------------------------
# *args with Regular Parameters
# ---------------------------------------------------------------------------

def make_pizza(size, *toppings):
    """Make pizza with any number of toppings"""
    print(f"Making a {size} pizza")
    print("Toppings:")
    for topping in toppings:
        print(f"  - {topping}")

make_pizza("Large", "Cheese", "Pepperoni", "Mushrooms")
make_pizza("Medium", "Cheese")
```

```python
# ---------------------------------------------------------------------------
# **kwargs - Variable Keyword Arguments
# ---------------------------------------------------------------------------

def print_user_info(**user_details):
    """Print user information from keyword arguments"""
    print(f"Type: {type(user_details)}")  # dict
    print("User Information:")
    for key, value in user_details.items():
        print(f"  {key}: {value}")

print_user_info(name="Alice", age=25, city="Boston")
print_user_info(name="Bob", profession="Engineer", country="USA", hobby="Gaming")
```

```python
# ---------------------------------------------------------------------------
# Combining *args and **kwargs
# ---------------------------------------------------------------------------

def advanced_function(required, *args, default="default", **kwargs):
    """Demonstrate all parameter types together"""
    print(f"Required: {required}")
    print(f"Args: {args}")
    print(f"Default: {default}")
    print(f"Kwargs: {kwargs}")
    print("-" * 40)

advanced_function("Must have")
advanced_function("Must have", 1, 2, 3)
advanced_function("Must have", 1, 2, default="custom")
advanced_function("Must have", 1, 2, default="custom", extra1="value1", extra2="value2")
```

```python
# ---------------------------------------------------------------------------
# Real-World Use Case - Flexible Logger
# ---------------------------------------------------------------------------

def log_message(level, message, *tags, **metadata):
    """Flexible logging function"""
    print(f"[{level}] {message}")
    
    if tags:
        print(f"Tags: {', '.join(tags)}")
    
    if metadata:
        print("Metadata:")
        for key, value in metadata.items():
            print(f"  {key}: {value}")
    print("-" * 40)

log_message("INFO", "User logged in")
log_message("WARNING", "High memory usage", "performance", "memory")
log_message("ERROR", "Database connection failed", "database", "critical",
            user_id=123, ip_address="192.168.1.1", retry_count=3)
```

```python
# ---------------------------------------------------------------------------
# UNPACKING ARGUMENTS
# ---------------------------------------------------------------------------

"""
ARGUMENT UNPACKING:
-------------------
Using * and ** to unpack sequences and dictionaries as function arguments.

* operator: Unpacks sequences (lists, tuples) into positional arguments
** operator: Unpacks dictionaries into keyword arguments
"""

def calculate_total(price, quantity, discount):
    """Calculate total after discount"""
    subtotal = price * quantity
    total = subtotal - (subtotal * discount / 100)
    return total

# Pack values in a list
order_data = [100, 5, 10]  # price, quantity, discount

# Unpack list as arguments
total = calculate_total(*order_data)
print(f"Total: ${total}")
```

```python
# ---------------------------------------------------------------------------
# Example 35: Unpacking Tuples
# ---------------------------------------------------------------------------

def display_date(year, month, day):
    """Display formatted date"""
    return f"{day:02d}/{month:02d}/{year}"

date_tuple = (2024, 12, 25)
print(display_date(*date_tuple))
```

```python
# ---------------------------------------------------------------------------
# Unpacking Dictionaries with **
# ---------------------------------------------------------------------------

def create_user(username, email, age, country):
    """Create user with provided details"""
    return {
        'username': username,
        'email': email,
        'age': age,
        'country': country,
        'created': 'Today'
    }

# Pack values in dictionary
user_data = {
    'username': 'johndoe',
    'email': 'john@example.com',
    'age': 30,
    'country': 'USA'
}

# Unpack dictionary as keyword arguments
user = create_user(**user_data)
print(user)
```

```python
# ---------------------------------------------------------------------------
# Combining Unpacking Techniques
# ---------------------------------------------------------------------------

def complex_order(product, quantity, *features, **customer_info):
    """Process complex order with multiple argument types"""
    print(f"Product: {product}")
    print(f"Quantity: {quantity}")
    print(f"Features: {features}")
    print(f"Customer: {customer_info}")

# Using unpacking
base_args = ["Laptop", 2]
extra_features = ["Extended Warranty", "Gift Wrap", "Express Shipping"]
customer = {"name": "Alice", "email": "alice@example.com", "vip": True}

complex_order(*base_args, *extra_features, **customer)

```

```python
# ---------------------------------------------------------------------------
# Example 38: Unpacking in Function Calls - Practical Example
# ---------------------------------------------------------------------------

def send_email(to, subject, body, cc=None, bcc=None):
    """Send email with required and optional parameters"""
    print(f"To: {to}")
    print(f"Subject: {subject}")
    print(f"Body: {body}")
    if cc:
        print(f"CC: {cc}")
    if bcc:
        print(f"BCC: {bcc}")

# Email data from database/API
email_data = {
    'to': 'user@example.com',
    'subject': 'Welcome!',
    'body': 'Thanks for signing up',
    'cc': 'manager@example.com'
}

send_email(**email_data)
```

```python
# ---------------------------------------------------------------------------
# Example 39: Merging Dictionaries with Unpacking
# ---------------------------------------------------------------------------

def configure_app(**settings):
    """Configure application with settings"""
    print("Application Settings:")
    for key, value in settings.items():
        print(f"  {key}: {value}")

# Merge multiple configuration sources
default_config = {'debug': False, 'port': 8000}
user_config = {'port1': 3000, 'host': 'localhost'}
environment_config = {'database': 'postgresql'}

# Unpack and merge all configs
configure_app(**default_config, **user_config, **environment_config)
```
