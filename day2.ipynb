```python
#Histogram
'''shows the distribution of numerical data.'''
```

```python
import matplotlib.pyplot as plt
data=[45,50,52,55,60,62,65,67,70,72,74,76,78,80,82,
     85,87,90,92,95]
plt.hist(data)
plt.title("Histogram plot")
plt.xlabel("Range")
plt.ylabel('Frequency')
plt.show()
```

```python
#skewness
'''measures the asymmetry of data distribution
If data is perfectly balanced on both sides around mean---
no skewness
If data is tilted towards right ---Positive skewness
If data is tilted towards left --- Negative skewness'''
```

```python
'''No skewness:
Mean=Median=Mode
Positive skewness:
Mean>Median>Mode
Negative skewness:
Mean<Median<Mode'''
```

```python
'''Pearsons skewness coefficient:
3(Mean-Median)/standard deviation
Interpretation:
0 ------ Symmetrical
>0 ---- Positive skewness
<0 ---- Negative skewness'''
```

```python
import pandas as pd
data=[2,3,5,6,8,10,12,15,30]
df=pd.DataFrame(data,columns=['usage'])
print(df['usage'].skew())
```

```python
plt.hist(df['usage'],bins=5)
plt.show()
```

```python
print("Mean:",df['usage'].mean())
print("Median:",df['usage'].median())
```

```python
#Handling skewness:
#1.log transformation
'''applies logarithmic function to each value in dataset
x=log(x+1)'''
```

```python
import numpy as np
data=[20,25,28,30,35,40,45,50,200]
df=pd.DataFrame(data,columns=['income'])
df['log_income']=np.log(df['income']+1)
print(df)
```

```python
#Square root transformation:
'''x=sq.root of each data value'''
```

```python
data=[1,2,3,4,9,16,25]
df=pd.DataFrame(data,columns=['usage'])
df['sqrt_usage']=np.sqrt(df['usage'])
print(df)
```

```python
#Box-Cox Transformation
```

```python
from scipy.stats import boxcox
```

```python
data=[2,3,4,5,6,8,10,12,50]
df=pd.DataFrame(data,columns=['usage'])
transformed_data,lam_value=boxcox(df['usage'])
print("Lambda:",lam_value)
df['boxcox_usage']=transformed_data
print(df)
```

```python
#Handling Negative skewness:
#Reflection+log transformation
#Reflection + sq.root transformation
#Reflection+Box-cox transformation
#Power Transformation
```

```python
#Reflection method:
#Reflection formula: x=max(x)+1-x
```

```python
data=[2,30,32,35,37,40,42,45]
df=pd.DataFrame(data,columns=['values'])
max_value=df['values'].max()
df['reflected']=max_value+1-df['values']
df['log_transformed']=np.log(df['reflected'])
print(df)
```

```python
#Correlation
'''measures the strength and direction of the 
relationship between two variables.'''
```

```python
data=pd.DataFrame({'yoe':[2,4,6,8],
                  'sal':[50,60,70,80]})
```

```python
#corr
print(data.corr())
```

```python
plt.scatter(data['yoe'],data['sal'])
plt.xlabel("Years of experience")
plt.ylabel('Salary')
plt.title("correlation between years of experience and salary")
plt.show()
```
