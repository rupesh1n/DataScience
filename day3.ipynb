```python
#Inferential statistics
#parameter ---- numerical measure describing the population
#Eg:Average income of all employees in a company
```

```python
#statistic
'''numerical value calculated from a sample.
Eg:Average income of 100 selected employees.'''
```

```python
#Sampling
'''The process of selecting a subset from the
population.'''
```

```python
#Hypothesis
'''A hypothesis is a testable assumption or
statement about a population parameter based 
on sample data'''
```

```python
#Null hypothesis(H0)
'''represents:
no effect
no difference
status quo
The claim we try to reject
Eg:A company claims:
Average customer monthly data usage is 50GB'''
```

```python
#Alternative Hypothesis(Ha):
'''represents:
what we want to prove
opposite to null hypothesis'''
```

```python
'''A coca cola company claims 500 ml of coke in a bottle.
'''
```

```python
#Types of alternative hypothesis
#1.Two tailed test
'''Testing any difference'''
```

```python
#2.Right tailed test:
#testing greater than

#3.Left tailed test:
#testing less than
```

```python
'''A gym trainer claims:
Average daily steps of members is 10000 steps:
'''
```

```python
#p-value
'''probability value.
is the probability of observing the sample result 
assuming null hypothesis is true.
p-value tells us how strong the evidence is
against the null hypothesis.'''
```

```python
#Decision rule:
#significance level(alpha)=0.05
#if p<=alpha ---- reject null hypothesis
#if p>alpha ---- accept null hypothsis
```

```python
#Z-test
'''1.population standard deviation is known
2.sample size is large(n>=30) or data is normally
distributed.'''
```
