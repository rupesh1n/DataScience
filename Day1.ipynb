```python
#Variance
'''Measures how far each data point is from the mean.
Larger variance indicates more spread out data points.'''
```

```python
#Standard deviation
'''square root of variance.
It represents the average distance of each data point
from the mean.
small SD --- values close to mean
large SD --- values widely spread'''
```

```python
#percentile
'''40,50,60,70,80,90,100
If a student mark is at the 70th percentile---
70% of students scored less than or equal to that mark.
Percentage:represents score out of 100
Eg:80% marks

percentile:represents position in dataset
Eg:80th percentile ranking.'''
```

```python
'''Steps:
1.Organize the data into an acsending order.
2.Calculate the percentile location i by:
      i=(p/100) X n
      p--- percentile of interest
      i---- percentile location
      n --- no.of data points
3. Determine the location by either a or b:
 a) If i is a whole no:,the pth percentile is 
 the avg of value at the ith location and value
 at i+1 th location
 b)If i is not a whole no:,the pth percentile value
 is located at the whole no:part of i+1'''
```

```python
import pandas as pd
import seaborn as sns
```

```python
data=pd.DataFrame({'MonthlyusageGb':[5,8,12,
                                    15,7,9,50,6,11,8],
                  'Monthlybill':[300,450,600,750,
                                400,500,2000,350,650,450],
                  'Tenuremonths':[24,18,10,6,30,20,
                                 2,36,12,22],
                  'Churn':['No','No','Yes','Yes',
                          'No','No','Yes','No','Yes','No']})
```

```python
data.describe()
```

```python
data['MonthlyusageGb'].mean()
```

```python
data['MonthlyusageGb'].median()
```

```python
data['MonthlyusageGb'].mode()
```

```python
#Mean>Median
```

```python
'''typicall usage is around 8-9GB
Light users consume 5-7GB
heavy users consume 12GB+
Extreme use is 50GB
Data usage varies widely among customers'''
```
