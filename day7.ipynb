```python
#construct a linear model that explains the 
#linear relationship a car's mileage(MPG) has
#with its attributes
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
```

```python
data=pd.read_csv('auto-mpg.csv')
data.head()
```

```python
data.info()
```

```python
data.shape
```

```python
data.drop('car name',axis=1,inplace=True)
```

```python
data.shape
```

```python
#1:america 2:europe 3:asia
data['origin']=data['origin'].replace({1:'america',
                                      2:'europe',
                                      3:'asia'})
```

```python
data.head()
```

```python
data=pd.get_dummies(data,columns=['origin'])
```

```python
data.info()
```

```python
h=pd.DataFrame(data.horsepower.str.isdigit())
```

```python
data[h['horsepower']==False]
```

```python
#Imputation
#mean imputation
#median imputation
#mode imputation
```

```python
data=data.replace('?',np.nan)
```

```python
replacing=lambda x:x.fillna(x.median())
data=data.apply(replacing,axis=0)
data['horsepower']=data['horsepower'].astype('float64')
```

```python
data.info()
```

```python
data.corr()
```

```python
x=data.drop('mpg',axis=1)
y=data['mpg']
```

```python
x.head()
```

```python
y
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,
                                               test_size=0.3,
                                               random_state=1)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
regression_model=LinearRegression()
```

```python
regression_model.fit(x_train,y_train)
```

```python
print(regression_model.coef_)
```

```python
for idx,col_name in enumerate(x_train.columns):
    print("The coefficient for {} is {}".format(
    col_name,regression_model.coef_[idx]))
```

```python
'''Each coefficient tells:
How much the target(mpg) changes when that feature increases
by 1 unit'''
```

```python
'''If cylinders increases by 1,mpg decreases by 0.39
Cars with more cylinders consume more fuel --- lower mileage'''
```

```python
#Multicollinearity
```

```python
'''Multicollinearity occurs when:
Two or more input features are highly correlated 
with each other'''
```

```python
regression_model.score(x_test,y_test)
```

```python
from sklearn.metrics import mean_squared_error
```

```python
y_pred=regression_model.predict(x_test)
mse=mean_squared_error(y_test,y_pred)
print("MSE=",mse)
rmse=np.sqrt(mse)
print("RMSE:",rmse)
```
