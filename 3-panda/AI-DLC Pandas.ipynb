A Series is a one-dimensional labeled array capable of holding any data type (integers, floats, strings, Python objects, etc.). Think of it as a single column from a spreadsheet, with an index attached to each value.

```python
import pandas as pd
import numpy as np
 
# From a list
s1 = pd.Series([10, 20, 30, 40])
print("s1",s1)
# 0    10
# 1    20   <- auto-generated integer index
# 2    30
# 3    40
 
# From a list with custom index
s2 = pd.Series([10, 20, 30], index=['a', 'b', 'c'])
print("s2",s2)

# From a dictionary (keys become index)
s3 = pd.Series({'Jan': 100, 'Feb': 200, 'Mar': 150})
print("s3",s3)

# From a NumPy array
s4 = pd.Series(np.array([1.1, 2.2, 3.3]), name='measurements')
print("s4",s4)
 
# Scalar broadcast
s5 = pd.Series(5, index=['a','b','c','d'])  # All values = 5
print("s5",s5)

# From a list with missing values
s6 = pd.Series([10, None, 30, np.nan], index=['a', 'b', 'c', 'd'])
print("s6",s6)   

# From a list with mixed types
s7 = pd.Series([10, 'twenty', 30.5, None], index=['a', 'b', 'c', 'd'])
print("s7",s7)   

# From a list with a datetime index
dates = pd.date_range('2024-01-01', periods=4)
s8 = pd.Series([100, 200, 150, 300], index=dates)
print("s8",s8)
```

#### Indexing and Slicing

```python
s = pd.Series([10, 20, 30, 40, 50], index=['a','b','c','d','e'])
 
# Label-based access
print(s['a'])          # 10
print(s[['a','c']])    # Multiple labels
 
# Position-based access
print(s.iloc[0])       # 10
print(s.iloc[1:3])     # b=20, c=30
 
# Boolean masking
print(s[s > 25])       # c=30, d=40, e=50
```

#### Vectorized Operations

```python
s = pd.Series([10, 20, 30, 40])
 
# Arithmetic — element-wise
s + 5           # 15, 25, 35, 45
s * 2           # 20, 40, 60, 80
s ** 2          # 100, 400, 900, 1600
 
# Operations between two Series — aligns by index
s1 = pd.Series({'a':1, 'b':2})
s2 = pd.Series({'b':10, 'c':20})
s1 + s2         # a=NaN, b=12, c=NaN  (index alignment)
 
# Common methods
# print each result to see the output

print("Sum:", s.sum())

print("Mean:", s.mean())

print("Cumulative Sum:", s.cumsum())

print("Description:\n", s.describe())    # Count, mean, std, min, percentiles, max
```

💡 Tip: Series alignment by index is one of the most powerful features of Pandas — it avoids many off-by-one errors common in raw Python/NumPy.

#### Data Indexing and Selection

Pandas offers multiple ways to select data. The two primary accessors are .loc[] for label-based and .iloc[] for position-based indexing. Understanding the difference is critical for correct data selection.
.loc[] — Label-Based Selection
Use .loc[] when you know the index labels (row names) and/or column names.


```python
import pandas as pd
#add few more rows to show loc examples
df = pd.DataFrame({
    'Name':['Alice','Bob','Raj','Charlie','Dave','Akash'],
    'Age':[25,30,35,28,40,22],
    'Score':[90,85,92,88,95,80]
}, index=['r1','r2','r3','r4','r5','r6'])
print(df)

# Single row
print(df.loc['r1'])                    # Series for row r1
 
# Slice rows (inclusive on both ends with loc)
print(df.loc['r1':'r3'])               # rows r1, r2, r3
 
# Row + column selection
print(df.loc['r1', 'Age'])             # 25
print(df.loc['r1':'r2', ['Name','Score']])
 
# All rows, specific columns
print(df.loc[:, 'Age'])                # Same as df['Age']

```

.iloc[] — Integer Position-Based Selection
Use .iloc[] when you want to select by position (0-based), ignoring the index labels.


```python
# Single element
df.iloc[0, 1]        # Row 0, Col 1 → 25
 
# Rows slice (exclusive end like Python lists)
df.iloc[0:2]         # Rows 0 and 1 only
 
# Row + column slice
df.iloc[0:3, 1:3]   # Rows 0-2, Cols 1-2
 
# Last row
df.iloc[-1]
 
# Every other row
df.iloc[::2]
```

```python
df = pd.read_csv("sales.csv")

# Total sales per region
region_sales = df.groupby("region")["amount"].sum()

# Average order value per product category
avg_order = df.groupby("category")["amount"].mean()

# Multiple aggregations at once
summary = df.groupby("region").agg(
    total_sales   = ("amount", "sum"),
    avg_order     = ("amount", "mean"),
    num_orders    = ("order_id", "count")
)
print(summary)
```

#### Detecting Missing Data

Before fixing missing data, always measure how much is missing and where. Pandas provides several methods to detect NaN values across the sales DataFrame.

```python
import pandas as pd
import numpy as np
 
df = pd.read_csv('sales.csv', parse_dates=['date'])
 
# Check for missing values — returns True where NaN
df.isnull()
df.isna()             # Same as isnull()
df.notnull()          # True where NOT NaN
 
# Count missing per column
df.isnull().sum()
# date        0
# product     0
# quantity    13    <-- 13 rows have no quantity
# price        0
# region       2    <-- 2 rows have no region
 
# Percentage missing per column
(df.isnull().sum() / len(df) * 100).round(2)
# quantity    43.33
# region       6.67
 
# Any missing in the entire DataFrame?
df.isnull().any().any()    # True
 
# Heatmap visualization (with seaborn)
# import seaborn as sns
# sns.heatmap(df.isnull(), cbar=False, yticklabels=False)
```

💡 Tip: Check missingness patterns early. Here, quantity is missing in 43% of rows — too many to simply drop. Region is only missing in 2 rows, making it safe to drop or fill with a default.

#### Handling Missing Data
The right strategy depends on why data is missing. In our sales dataset, missing quantity may mean the sale was not completed, while missing region is likely a data entry error.


```python
import pandas as pd

# Clean column names first
df.columns = df.columns.str.strip()

# --- dropna() examples (use ONE at a time, not all together) ---
# df = df.dropna()                       # drops rows with any missing value
# df = df.dropna(thresh=4)               # keep rows with >=4 non-NaN
# df = df.dropna(subset=['region'])      # drop rows where region is missing

# --- fillna() ---
# Fill 'quantity' with mean/median only if column exists
if 'quantity' in df.columns:
    df['quantity'] = pd.to_numeric(df['quantity'], errors='coerce')  # ensure numeric
    df['quantity'] = df['quantity'].fillna(df['quantity'].mean())   # or median()

# Forward / backward fill (modern pandas)
df = df.ffill()
df = df.bfill()

# Fill each column differently (only applies to columns that exist)
fill_map = {}
if 'quantity' in df.columns:
    fill_map['quantity'] = df['quantity'].median()
if 'region' in df.columns:
    fill_map['region'] = 'Unknown'

if fill_map:
    df = df.fillna(fill_map)

print(df)
```

#### Removing Duplicates

```python
# Detect duplicates
df.duplicated()                              # True for fully duplicate rows
df.duplicated(subset=['date', 'product'])    # Same product sold twice on same date
 
# Count duplicates
df.duplicated().sum()                        # How many duplicate rows exist?
 
# Remove duplicates
df.drop_duplicates()                         # Keep first occurrence
df.drop_duplicates(keep='last')              # Keep last occurrence
df.drop_duplicates(keep=False)               # Drop ALL copies of any duplicate
df.drop_duplicates(subset=['date','product'])# Deduplicate by date + product
```

#### Data Type Conversion
After loading sales.csv, check that each column has the right dtype. Pandas may read quantity as float64 due to NaN values, and date as object (string) if not parsed on load.


```python
# View current dtypes
df.dtypes
# date        object   (if not parsed on load)
# product     object
# quantity    float64  (NaN forces float)
# price       int64
# region      object
 
# Convert date column to datetime
df['date'] = pd.to_datetime(df['date'])                    # Auto-detect format
df['date'] = pd.to_datetime(df['date'], format='%Y-%m-%d') # Explicit format
df['date'] = pd.to_datetime(df['date'], errors='coerce')   # Bad dates → NaT
 
# After filling NaN, convert quantity to int
df['quantity'] = df['quantity'].fillna(0).astype(int)
 
# Safer conversion with error handling
df['price'] = pd.to_numeric(df['price'], errors='coerce')  # Non-numeric → NaN
 
# Convert product and region to category (saves memory)
df['product'] = df['product'].astype('category')
df['region']  = df['region'].astype('category')
```
